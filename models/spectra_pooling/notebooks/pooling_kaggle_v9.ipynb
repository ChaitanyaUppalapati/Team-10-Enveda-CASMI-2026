# Member 7: CASMI on Kaggle, v9 — Architecture 3: contrastive peak transformer (offline, 2× T4)

**Where we were.** v7 (LB **0.281**): Member 3's ResidualMLP predicts a 2,048-bit fingerprint, candidates are
ranked by Tanimoto, a LambdaRank reranker re-orders the top of each list. After reranking, **70–80%** of the
validation MRR still lost is the answer beaten by a **same-formula isomer**: close isomers share almost every
fingerprint bit, so "guess the bits, then compare" cannot separate them. v8 tried a documentation prior for this
(LB 0.276): validation answers are well-studied library standards, so it rewarded fame, not chemistry.

**Architecture 3** learns the comparison itself (CLIP-style, `common/contrastive.py`):

| Part | v7 | v9 |
|---|---|---|
| spectrum | 1-Da bins → MLP | **peak tokens** (exact m/z and neutral loss as Fourier features) → **transformer** |
| a molecule's spectra | attention / mean over frozen fingerprints | **gated attention trained with the encoder** (starts as the mean) |
| candidate | RDKit bits, Tanimoto | **molecule encoder** (MLP on the bits) |
| objective | BCE per bit | **InfoNCE**: the right structure vs the batch **and same-mass hard negatives (isomers)** |

It enters the pipeline three ways, each measured on the same split, pool and reranker: as its own first stage,
fused with the MLP's first stage, and as reranker features (`clip_*`, own ablation step).

**Synthetic-library penalty (test only).** The test comes from biological extracts; `enveda-180` (182,941 of the
275,810 training structures) is a synthetic, drug-like library. Inside each test top 25, nothing dropped, order
otherwise kept: **tier A** = enveda-180 structures that are not natural products (not COCONUT, not an NP example)
go to the end, rank 1 included; **tier B** = other training structures that are not natural products go after the
unflagged ones on ranks 2-25 (rank 1 is never moved for tier B). Validation cannot tune this (its answers include
those structures), so it is a fixed rule; the two tiers follow the public CASMI 2026 notebooks, the code is ours.

**Off in v9:** the documentation prior and the PubChem channel (v8). **Attach (4):** competition data,
`rdkit_cp313`, `coconut-2026-11-lite`, `pubchemlite-2026-09`. T4 ×2, Internet **off**. ~3–3.5 h.

```python
%%writefile m7/casmi_pipeline.py
"""Retrieval and ranking pipeline for molecule identification from MS/MS spectra.

Provides data loading, validation splitting, candidate retrieval, spectral library
search, neural fingerprint predictors, MRR@25 evaluation, submission writing, run
logging and figures for the Enveda CASMI 2026 competition.
"""

import glob
import hashlib
import os
import re
import sys
import time

import numpy as np
import pandas as pd
import pyarrow.parquet as pq

PROTON = 1.00727646688
ELECTRON = 0.00054857990907
TOP_K = 25
MAX_PPM_LABEL_ERROR = 20.0  # library spectra with a larger precursor error are treated as bad labels

# adduct -> (molecules per ion, absolute charge, mass added to the n*M term)
ADDUCTS = {
    "[M+H]+": (1, 1, PROTON),
    "[M-H]-": (1, 1, -PROTON),
    "[M+Na]+": (1, 1, 22.98976928 - ELECTRON),
    "[M+K]+": (1, 1, 38.96370649 - ELECTRON),
    "[M+NH4]+": (1, 1, 18.03437413 - ELECTRON),
    "[M+Cl]-": (1, 1, 34.96885268 + ELECTRON),
    "[M+CH2O2-H]-": (1, 1, 46.00547930 - PROTON),
    "[M+C2H4O2-H]-": (1, 1, 60.02112937 - PROTON),
    "[M-H2O+H]+": (1, 1, PROTON - 18.01056468),
    "[M-2H2O+H]+": (1, 1, PROTON - 36.02112937),
    "[M]+": (1, 1, -ELECTRON),
    "[M]-": (1, 1, ELECTRON),
    "[M+2H]2+": (1, 2, 2 * PROTON),
    "[2M+H]+": (2, 1, PROTON),
    "[2M+Na]+": (2, 1, 22.98976928 - ELECTRON),
    "[2M-H]-": (2, 1, -PROTON),
}
QUERY_ADDUCTS = ["[M+H]+", "[M-H]-", "[M+Na]+", "[M+NH4]+", "[M+K]+", "[M+Cl]-", "[M+CH2O2-H]-"]
ADDUCT_INDEX = {a: i + 1 for i, a in enumerate(ADDUCTS)}  # 0 means "other"
N_ADDUCTS = len(ADDUCTS) + 1

MONO = {
    "H": 1.00782503207,
    "B": 11.0093054,
    "C": 12.0,
    "N": 14.0030740048,
    "O": 15.99491461956,
    "F": 18.99840322,
    "Na": 22.9897692809,
    "Mg": 23.9850417,
    "Al": 26.98153863,
    "Si": 27.9769265325,
    "P": 30.97376163,
    "S": 31.972071,
    "Cl": 34.96885268,
    "K": 38.96370668,
    "Ca": 39.96259098,
    "Fe": 55.9349375,
    "Co": 58.933195,
    "Ni": 57.9353429,
    "Cu": 62.9295975,
    "Zn": 63.9291422,
    "As": 74.9215965,
    "Se": 79.9165213,
    "Br": 78.9183371,
    "Sn": 119.9021947,
    "I": 126.904473,
    "Pt": 194.9647911,
    "Hg": 201.970643,
}


def find_data_dir(explicit=None):
    """Return the folder holding train.parquet on Kaggle, Colab or a local machine."""
    if explicit and os.path.exists(os.path.join(explicit, "train.parquet")):
        return explicit
    for root in [
        "/kaggle/input/**",
        "/content/casmi_data",
        "../enveda-CASMI26-molecule-id-mass-spectra",
        "../../enveda-CASMI26-molecule-id-mass-spectra",
        ".",
    ]:
        hits = glob.glob(os.path.join(root, "train.parquet"), recursive=True)
        if hits:
            return os.path.dirname(hits[0])
    raise FileNotFoundError("train.parquet not found. Check that the competition data is available to this runtime.")


def formula_mass(formula):
    total = 0.0
    for element, count in re.findall(r"([A-Z][a-z]?)(\d*)", str(formula)):
        if element not in MONO:
            return np.nan
        total += MONO[element] * (int(count) if count else 1)
    return total if total > 0 else np.nan


def neutral_mass(precursor_mz, adduct, mode=None):
    """Neutral monoisotopic mass of one spectrum. Unknown adducts fall back to +/- one proton."""
    if adduct in ADDUCTS:
        n, z, delta = ADDUCTS[adduct]
    else:
        n, z, delta = 1, 1, (-PROTON if str(mode).startswith("neg") else PROTON)
    return (precursor_mz * z - delta) / n


def add_neutral_mass(df):
    df = df.copy()
    df["neutral_mass"] = [
        neutral_mass(m, a, p) for m, a, p in zip(df.precursor_mz, df.adduct, df.ionization_mode, strict=False)
    ]
    return df


def load_meta(data_dir):
    """Load everything in train.parquet except the peak lists."""
    cols = [
        "ingest_lib",
        "inchikey14",
        "normalized_smiles",
        "molecular_formula",
        "ionization_mode",
        "instrument_type",
        "adduct",
        "precursor_mz",
        "precursor_error_ppm",
        "num_peaks",
    ]
    meta = pd.read_parquet(os.path.join(data_dir, "train.parquet"), columns=cols)
    meta["row_id"] = np.arange(len(meta))
    for col in ["ingest_lib", "ionization_mode", "instrument_type", "adduct"]:
        meta[col] = meta[col].astype("category")
    meta["good_label"] = meta.precursor_error_ppm.abs() <= MAX_PPM_LABEL_ERROR
    return meta


def _unit(key, salt):
    digest = hashlib.md5(f"{salt}:{key}".encode()).hexdigest()
    return int(digest[:12], 16) / float(16**12)


def make_split(meta, frac=0.10, salt="casmi-split-v1", max_query_spectra=6):
    """Structure-level validation split that simulates test classes 1 and 2.

    Class 1: the structure keeps reference spectra, but only from other libraries.
    The whole library the query comes from is removed for that structure, because
    the real test spectra are unpublished.
    Class 2: every spectrum of the structure is removed. The structure stays in the
    candidate pool with no reference spectra.

    Returns (queries, library_mask). Queries has one row per query spectrum.
    """
    keys = meta.inchikey14.unique()
    np_keys = set(meta.loc[meta.ingest_lib == "enveda-np-examples", "inchikey14"].unique())
    held = {k for k in keys if _unit(k, salt) < frac or k in np_keys}
    sub = meta[meta.inchikey14.isin(held)]
    excluded = np.zeros(len(meta), dtype=bool)
    rows = []
    rng = np.random.default_rng(20261110)
    for key, group in sub.groupby("inchikey14", sort=True, observed=True):
        eligible = group[group.good_label & group.adduct.isin(QUERY_ADDUCTS) & (group.num_peaks >= 4)]
        if eligible.empty:
            excluded[group.row_id.to_numpy()] = True
            continue
        libs = eligible.ingest_lib.astype(str)
        counts = libs.value_counts()
        if "enveda-np-examples" in counts.index:
            query_lib = "enveda-np-examples"
        elif "enveda-180" in counts.index:
            query_lib = "enveda-180"
        else:
            query_lib = counts.index[0]
        others = group[(group.ingest_lib.astype(str) != query_lib) & group.good_label]
        cls = 1 if (_unit(key, salt + "-class") < 0.5 and len(others) > 0) else 2
        if cls == 1:
            excluded[group.loc[group.ingest_lib.astype(str) == query_lib, "row_id"].to_numpy()] = True
        else:
            excluded[group.row_id.to_numpy()] = True
        pool_rows = eligible[libs == query_lib]
        n = int(min(len(pool_rows), rng.integers(1, max_query_spectra + 1)))
        picked = rng.choice(pool_rows.row_id.to_numpy(), size=n, replace=False)
        for row_id in picked:
            rows.append((int(row_id), f"val_{key}", key, cls, query_lib))
    queries = pd.DataFrame(rows, columns=["row_id", "molecule_id", "true_key", "cls", "query_lib"])
    info = meta.iloc[queries.row_id.to_numpy()]
    for col in ["precursor_mz", "adduct", "ionization_mode"]:
        queries[col] = info[col].astype(object if col != "precursor_mz" else float).to_numpy()
    queries = add_neutral_mass(queries)
    return queries, ~excluded


def sample_molecules(queries, n, seed=7):
    """Keep the enveda-np-examples proxy set, then sample the rest evenly across classes 1 and 2."""
    per_mol = queries.drop_duplicates("molecule_id")
    is_proxy = per_mol.query_lib == "enveda-np-examples"
    keep = set(per_mol[is_proxy].molecule_id)
    rest = per_mol[~is_proxy]
    per_class = max((n - len(keep)) // 2, 0)
    for cls in (1, 2):
        part = rest[rest.cls == cls].molecule_id
        keep |= set(part.sample(min(len(part), per_class), random_state=seed))
    return queries[queries.molecule_id.isin(keep)].reset_index(drop=True)


def build_pool(meta, library_mask, extra_candidates=None):
    """One row per structure, sorted by mass. n_ref counts usable library spectra."""
    pool = meta.drop_duplicates("inchikey14")[["inchikey14", "normalized_smiles", "molecular_formula"]].copy()
    pool.columns = ["key", "smiles", "formula"]
    masses = {f: formula_mass(f) for f in pool.formula.unique()}
    pool["mass"] = pool.formula.map(masses)
    counts = meta.loc[library_mask & meta.good_label.to_numpy(), "inchikey14"].value_counts()
    pool["n_ref"] = pool.key.map(counts).fillna(0).astype(int)
    if extra_candidates is not None:  # optional external candidates with columns: key, smiles, mass
        extra = extra_candidates[~extra_candidates.key.isin(set(pool.key))].copy()
        extra["formula"], extra["n_ref"] = "", 0
        pool = pd.concat([pool, extra[pool.columns]], ignore_index=True)
    pool = pool.dropna(subset=["mass", "smiles"])
    return pool.sort_values(["mass", "key"], kind="stable").reset_index(drop=True)


def consensus_masses(spectra):
    """Median neutral mass per molecule across its spectra and adducts."""
    return spectra.groupby("molecule_id", sort=False).neutral_mass.median()


def window_indices(pool_mass, mass, ppm):
    tol = mass * ppm * 1e-6
    return np.arange(np.searchsorted(pool_mass, mass - tol), np.searchsorted(pool_mass, mass + tol, side="right"))


def nearest_indices(pool_mass, mass, n):
    """The n pool entries closest in mass. Used to fill short lists."""
    center = int(np.searchsorted(pool_mass, mass))
    lo, hi = max(center - n, 0), min(center + n, len(pool_mass))
    idx = np.arange(lo, hi)
    return idx[np.argsort(np.abs(pool_mass[idx] - mass), kind="stable")[:n]]


def fill_to_k(ranked, pool_mass, mass, k=TOP_K):
    """Append nearest-mass structures so the list has k unique entries."""
    ranked = list(dict.fromkeys(int(i) for i in ranked))[:k]
    if len(ranked) < k:
        seen = set(ranked)
        for i in nearest_indices(pool_mass, mass, 4 * k):
            if int(i) not in seen:
                ranked.append(int(i))
                seen.add(int(i))
            if len(ranked) == k:
                break
    return ranked


def rank_b1(pool, mass, ppm=10.0, order="ppm_then_prior"):
    """B1: rank by mass alone. Isomers tie on mass, so reference count breaks ties."""
    pool_mass = pool.mass.to_numpy()
    idx = window_indices(pool_mass, mass, ppm)
    if len(idx):
        err = np.round(np.abs(pool_mass[idx] - mass) / mass * 1e6, 3)
        n_ref = pool.n_ref.to_numpy()[idx]
        primary = (err, -n_ref) if order == "ppm_then_prior" else (-n_ref, err)
        idx = idx[np.lexsort((idx,) + primary[::-1])]
    return fill_to_k(idx, pool_mass, mass)


def rank_table(rankings, truth, pool_keys):
    """One row per molecule with the rank of its true structure (0 means missed)."""
    records = []
    for row in truth.itertuples():
        keys = [pool_keys[i] for i in rankings[row.molecule_id][:TOP_K]]
        rank = keys.index(row.true_key) + 1 if row.true_key in keys else 0
        records.append(
            (row.molecule_id, row.cls, row.query_lib == "enveda-np-examples", 1.0 / rank if rank else 0.0, rank)
        )
    return pd.DataFrame(records, columns=["molecule_id", "cls", "np_proxy", "rr", "rank"])


def evaluate(rankings, truth, pool_keys, ci=False):
    """Per-class MRR@25 and hit rates. 'balanced' averages classes 1 and 2, since the real mix is hidden."""
    res = rank_table(rankings, truth, pool_keys)
    rng = np.random.default_rng(3)

    def summarize(part):
        out = {
            "n": len(part),
            "MRR@25": round(part.rr.mean(), 4),
            "top1": round((part["rank"] == 1).mean(), 4),
            "top5": round(part["rank"].between(1, 5).mean(), 4),
            "top25": round((part["rank"] > 0).mean(), 4),
        }
        if ci and len(part):
            draws = rng.choice(part.rr.to_numpy(), size=(1000, len(part))).mean(axis=1)
            out["MRR_ci_low"], out["MRR_ci_high"] = (
                round(np.quantile(draws, 0.025), 4),
                round(np.quantile(draws, 0.975), 4),
            )
        return out

    table = pd.DataFrame(
        {
            "class 1": summarize(res[res.cls == 1]),
            "class 2": summarize(res[res.cls == 2]),
            "np proxy": summarize(res[res.np_proxy]),
        }
    ).T
    balanced = table.loc[["class 1", "class 2"]].mean().round(4)
    balanced["n"] = table.loc[["class 1", "class 2"], "n"].sum()
    table.loc["balanced"] = balanced
    return table


def truth_table(queries):
    return queries.drop_duplicates("molecule_id")[["molecule_id", "true_key", "cls", "query_lib"]]


# ---------- spectra ----------


def clean_spectrum(mz, intensity, precursor, max_peaks=100, floor=0.005):
    """Drop the precursor region, apply an intensity floor, cap the peak count and sqrt-scale intensities."""
    keep = (mz > 0) & (intensity > 0) & (mz < precursor - 0.5)
    mz, intensity = mz[keep], intensity[keep]
    if len(mz) == 0:
        return mz, intensity
    keep = intensity >= floor * intensity.max()
    mz, intensity = mz[keep], intensity[keep]
    if len(mz) > max_peaks:
        top = np.argpartition(intensity, -max_peaks)[-max_peaks:]
        mz, intensity = mz[top], intensity[top]
    order = np.argsort(mz)
    mz, intensity = mz[order], np.sqrt(intensity[order])
    return mz, intensity / np.linalg.norm(intensity)


def read_spectra(parquet_path, row_ids, max_peaks=100):
    """Read and clean selected rows into padded arrays, in the order of row_ids."""
    row_ids = np.asarray(row_ids, dtype=np.int64)
    out_mz = np.zeros((len(row_ids), max_peaks), dtype=np.float32)
    out_in = np.zeros((len(row_ids), max_peaks), dtype=np.float32)
    order = np.argsort(row_ids, kind="stable")
    sorted_ids = row_ids[order]
    file = pq.ParquetFile(parquet_path)
    start = 0
    for group in range(file.metadata.num_row_groups):
        n_rows = file.metadata.row_group(group).num_rows
        lo, hi = np.searchsorted(sorted_ids, [start, start + n_rows])
        if hi > lo:
            table = file.read_row_group(group, columns=["ms2_mzs", "ms2_normalized_intensities", "precursor_mz"])
            mz_col = table["ms2_mzs"].combine_chunks()
            in_col = table["ms2_normalized_intensities"].combine_chunks()
            mz_off, in_off = mz_col.offsets.to_numpy(), in_col.offsets.to_numpy()
            mz_val, in_val = mz_col.values.to_numpy(), in_col.values.to_numpy()
            precursor = table["precursor_mz"].to_numpy()
            for position in range(lo, hi):
                local = sorted_ids[position] - start
                mz, intensity = clean_spectrum(
                    mz_val[mz_off[local] : mz_off[local + 1]],
                    in_val[in_off[local] : in_off[local + 1]],
                    precursor[local],
                    max_peaks,
                )
                target = order[position]
                out_mz[target, : len(mz)] = mz
                out_in[target, : len(mz)] = intensity
            del table, mz_col, in_col, mz_val, in_val
        start += n_rows
    return out_mz, out_in


def load_test(data_dir):
    path = os.path.join(data_dir, "test.parquet")
    test = add_neutral_mass(
        pd.read_parquet(path, columns=["molecule_id", "spectrum_id", "adduct", "ionization_mode", "precursor_mz"])
    )
    test["row_id"] = np.arange(len(test))
    return test, path


def match_score(q_mz, q_in, l_mz, l_in, tol=0.01, shift=None, min_matches=3):
    """Greedy cosine. With shift set, peaks may also match after a precursor shift (modified cosine)."""
    diff = q_mz[:, None] - l_mz[None, :]
    mask = np.abs(diff) <= tol
    if shift is not None and abs(shift) > tol:
        mask |= np.abs(diff - shift) <= tol
    i, j = np.nonzero(mask)
    if len(i) < min_matches:
        return 0.0
    products = q_in[i] * l_in[j]
    used_q, used_l, total, matches = set(), set(), 0.0, 0
    for k in np.argsort(-products, kind="stable"):
        a, b = int(i[k]), int(j[k])
        if a not in used_q and b not in used_l:
            used_q.add(a)
            used_l.add(b)
            total += float(products[k])
            matches += 1
    return total if matches >= min_matches else 0.0


class SpectralLibrary:
    """Cleaned library spectra, restricted to structures near the query masses."""

    def __init__(self, data_dir, meta, library_mask, pool, query_masses, ppm=10.0, per_candidate=30, max_peaks=100):
        pool_mass = pool.mass.to_numpy()
        needed = np.zeros(len(pool), dtype=bool)
        for mass in query_masses:
            needed[window_indices(pool_mass, mass, ppm)] = True
        wanted = set(pool.key.to_numpy()[needed])
        usable = meta[
            library_mask
            & meta.good_label.to_numpy()
            & meta.inchikey14.isin(wanted).to_numpy()
            & (meta.num_peaks >= 3).to_numpy()
        ].copy()
        # Prefer the test instrument and simple adducts when a structure has many spectra.
        usable["priority"] = (usable.instrument_type.astype(str) != "timsTOF").astype(int) + (
            ~usable.adduct.isin(QUERY_ADDUCTS)
        ).astype(int)
        usable = usable.sort_values(["inchikey14", "ionization_mode", "priority", "row_id"], kind="stable")
        usable = (
            usable.groupby(["inchikey14", "ionization_mode"], observed=True).head(per_candidate).reset_index(drop=True)
        )
        self.mz, self.intensity = read_spectra(
            os.path.join(data_dir, "train.parquet"), usable.row_id.to_numpy(), max_peaks
        )
        self.n_peaks = (self.intensity > 0).sum(axis=1)
        self.precursor = usable.precursor_mz.to_numpy()
        self.index = {}
        for position, (key, mode) in enumerate(
            zip(usable.inchikey14.to_numpy(), usable.ionization_mode.astype(str).to_numpy(), strict=False)
        ):
            self.index.setdefault((key, mode), []).append(position)


def rank_b2(pool, library, spectra_mz, spectra_in, group, ppm=10.0, modified=True, pooling="max"):
    """B2: mass window, then spectral similarity pooled across the molecule's spectra.

    group is the molecule's rows, positionally aligned with spectra_mz/spectra_in via column pos.
    """
    pool_mass = pool.mass.to_numpy()
    keys = pool.key.to_numpy()
    n_ref = pool.n_ref.to_numpy()
    mass = float(group.neutral_mass.median())
    idx = window_indices(pool_mass, mass, ppm)
    if len(idx) == 0:
        return fill_to_k([], pool_mass, mass)
    scores = np.zeros((len(group), len(idx)))
    for s, row in enumerate(group.itertuples()):
        n = int((spectra_in[row.pos] > 0).sum())
        if n == 0:
            continue
        q_mz, q_in = spectra_mz[row.pos, :n], spectra_in[row.pos, :n]
        for c, pool_index in enumerate(idx):
            best = 0.0
            for lib_pos in library.index.get((keys[pool_index], row.ionization_mode), ()):
                m = library.n_peaks[lib_pos]
                if m == 0:
                    continue
                shift = (row.precursor_mz - library.precursor[lib_pos]) if modified else None
                value = match_score(q_mz, q_in, library.mz[lib_pos, :m], library.intensity[lib_pos, :m], shift=shift)
                best = max(best, value)
            scores[s, c] = best
    if pooling == "max":
        pooled = scores.max(axis=0)
    elif pooling == "mean":
        pooled = scores.mean(axis=0)
    elif pooling == "rrf":
        ranks = (-scores).argsort(axis=1, kind="stable").argsort(axis=1) + 1
        pooled = np.where(scores > 0, 1.0 / (60 + ranks), 0.0).sum(axis=0)
    else:
        raise ValueError(pooling)
    err = np.round(np.abs(pool_mass[idx] - mass) / mass * 1e6, 3)
    order = np.lexsort((idx, -n_ref[idx], err, -np.round(pooled, 6)))
    return fill_to_k(idx[order], pool_mass, mass)


def run_b2(pool, library, spectra, spectra_mz, spectra_in, **kwargs):
    spectra = spectra.copy()
    spectra["pos"] = np.arange(len(spectra))
    return {
        mol: rank_b2(pool, library, spectra_mz, spectra_in, group, **kwargs)
        for mol, group in spectra.groupby("molecule_id", sort=False)
    }


def write_submission(rankings, pool, test, path="submission.csv"):
    """Write submission.csv and enforce the competition's rejection rules."""
    smiles = pool.smiles.to_numpy()
    rows = []
    for mol in test.molecule_id.drop_duplicates():
        guesses = [smiles[i] for i in rankings[mol][:TOP_K]]
        if not guesses:
            raise ValueError(f"no guesses for {mol}")
        rows.append((mol, ";".join(guesses)))
    sub = pd.DataFrame(rows, columns=["molecule_id", "smiles"])
    assert len(sub) > 0 and sub.notna().all().all() and sub.molecule_id.is_unique
    assert (sub.smiles.str.count(";") < TOP_K).all() and (sub.smiles.str.len() > 0).all()
    sub.to_csv(path, index=False)
    return sub


def log_line(version, change, table, lb_score=""):
    """Row for the team experiment log: date, version, change, val MRR, LB score."""
    return {
        "date": time.strftime("%Y-%m-%d"),
        "version": version,
        "change": change,
        "val_MRR@25": float(table.loc["balanced", "MRR@25"]),
        "class1": float(table.loc["class 1", "MRR@25"]),
        "class2": float(table.loc["class 2", "MRR@25"]),
        "np_proxy": float(table.loc["np proxy", "MRR@25"]),
        "LB": lb_score,
    }


# ---------- neural fingerprint predictors ----------

FP_BITS = 2048


def morgan_packed(smiles_list, n_bits=FP_BITS, radius=2):
    """Packed Morgan fingerprints. Rows for unparsable SMILES stay all zero."""
    from rdkit import Chem, RDLogger
    from rdkit.Chem import rdFingerprintGenerator

    RDLogger.DisableLog("rdApp.*")
    generator = rdFingerprintGenerator.GetMorganGenerator(radius=radius, fpSize=n_bits)
    out = np.zeros((len(smiles_list), n_bits // 8), dtype=np.uint8)
    for i, smiles in enumerate(smiles_list):
        mol = Chem.MolFromSmiles(smiles)
        if mol is not None:
            out[i] = np.packbits(generator.GetFingerprintAsNumPy(mol).astype(bool))
    return out


def select_train_rows(meta, library_mask, per_structure=8, seed=11):
    """Cap spectra per structure so heavily measured compounds don't dominate training."""
    rows = meta[library_mask & meta.good_label.to_numpy() & (meta.num_peaks >= 3).to_numpy()]
    shuffled = rows.sample(frac=1.0, random_state=seed)
    return shuffled.groupby("inchikey14", observed=True).head(per_structure).sort_values("row_id")


def build_models():
    """Define the models inside a function so that the baselines do not require PyTorch."""
    import torch
    from torch import nn

    class ResidualBlock(nn.Module):
        def __init__(self, width, dropout):
            super().__init__()
            self.net = nn.Sequential(
                nn.LayerNorm(width), nn.Linear(width, width), nn.GELU(), nn.Dropout(dropout), nn.Linear(width, width)
            )

        def forward(self, x):
            return x + self.net(x)

    class ResidualMLP(nn.Module):
        """1-Da binned spectrum plus precursor and adduct, through residual blocks."""

        def __init__(self, n_bins=1200, width=2048, depth=4, dropout=0.2, n_adducts=N_ADDUCTS):
            super().__init__()
            self.n_bins = n_bins
            self.adduct = nn.Embedding(n_adducts, 32)
            self.stem = nn.Linear(n_bins + 33, width)
            self.blocks = nn.Sequential(*[ResidualBlock(width, dropout) for _ in range(depth)])
            self.head = nn.Sequential(nn.LayerNorm(width), nn.Linear(width, FP_BITS))

        def forward(self, mz, intensity, precursor, adduct):
            bins = mz.long().clamp(0, self.n_bins - 1)
            x = torch.zeros(mz.size(0), self.n_bins, device=mz.device, dtype=intensity.dtype)
            x.scatter_add_(1, bins, intensity)
            x = torch.cat([x, (precursor / 1000.0).unsqueeze(1), self.adduct(adduct)], dim=1)
            return self.head(self.blocks(self.stem(x)))

    class PeakSetTransformer(nn.Module):
        """Each peak is a token. No positional encoding, so peak order doesn't matter."""

        def __init__(self, d_model=256, layers=4, heads=8, dropout=0.1, n_freq=32, n_adducts=N_ADDUCTS):
            super().__init__()
            freqs = torch.logspace(-2.0, 3.0, n_freq) * 2 * torch.pi / 1000.0
            self.register_buffer("freqs", freqs)
            self.peak = nn.Sequential(nn.Linear(4 * n_freq + 1, d_model), nn.GELU(), nn.Linear(d_model, d_model))
            self.precursor = nn.Sequential(nn.Linear(2 * n_freq, d_model), nn.GELU(), nn.Linear(d_model, d_model))
            self.adduct = nn.Embedding(n_adducts, d_model)
            layer = nn.TransformerEncoderLayer(
                d_model, heads, 4 * d_model, dropout, activation="gelu", batch_first=True, norm_first=True
            )
            self.encoder = nn.TransformerEncoder(layer, layers, enable_nested_tensor=False)
            self.head = nn.Sequential(nn.LayerNorm(d_model), nn.Linear(d_model, FP_BITS))

        def fourier(self, value):
            angle = value.unsqueeze(-1) * self.freqs
            return torch.cat([angle.sin(), angle.cos()], dim=-1)

        def forward(self, mz, intensity, precursor, adduct):
            loss = precursor.unsqueeze(1) - mz  # neutral loss of each peak
            tokens = self.peak(torch.cat([self.fourier(mz), self.fourier(loss), intensity.unsqueeze(-1)], dim=-1))
            summary = (self.precursor(self.fourier(precursor)) + self.adduct(adduct)).unsqueeze(1)
            padding = torch.cat([torch.zeros_like(intensity[:, :1], dtype=torch.bool), intensity <= 0], dim=1)
            encoded = self.encoder(torch.cat([summary, tokens], dim=1), src_key_padding_mask=padding)
            return self.head(encoded[:, 0])

    return {"residual_mlp": ResidualMLP, "peak_set_transformer": PeakSetTransformer}


def adduct_ids(values):
    return np.array([ADDUCT_INDEX.get(a, 0) for a in values], dtype=np.int64)


def predict_probs(model, mz, intensity, precursor, adduct, device, batch=512):
    import torch

    model.eval()
    chunks = []
    with torch.no_grad():
        for i in range(0, len(mz), batch):
            sl = slice(i, i + batch)
            logits = model(
                torch.as_tensor(mz[sl], device=device),
                torch.as_tensor(intensity[sl], device=device),
                torch.as_tensor(precursor[sl], dtype=torch.float32, device=device),
                torch.as_tensor(adduct[sl], device=device),
            )
            chunks.append(torch.sigmoid(logits.float()).cpu().numpy())
    return np.concatenate(chunks)


def rank_by_fingerprint(pool, pool_fp_packed, spectra, probs, ppm=10.0):
    """Average probabilities across a molecule's spectra and rank candidates by soft Tanimoto."""
    pool_mass = pool.mass.to_numpy()
    spectra = spectra.copy()
    spectra["pos"] = np.arange(len(spectra))
    rankings = {}
    for mol, group in spectra.groupby("molecule_id", sort=False):
        mass = float(group.neutral_mass.median())
        idx = window_indices(pool_mass, mass, ppm)
        if len(idx):
            p = probs[group.pos.to_numpy()].mean(axis=0)
            bits = np.unpackbits(pool_fp_packed[idx], axis=1).astype(np.float32)
            overlap = bits @ p
            tanimoto = overlap / (p.sum() + bits.sum(axis=1) - overlap + 1e-8)
            idx = idx[np.argsort(-tanimoto, kind="stable")]
        rankings[mol] = fill_to_k(idx, pool_mass, mass)
    return rankings


def train_model(
    model, data, val, device, epochs=10, batch=512, lr=3e-4, max_minutes=240, out_path="model.pt", log=print
):
    """Train with BCE and keep the checkpoint with the best validation MRR@25.

    data: dict with mz, intensity, precursor, adduct, target_index, fp_packed.
    val: dict with pool, pool_fp, queries, mz, intensity, truth.
    """
    import contextlib

    import torch
    from torch import nn

    model = model.to(device)
    tensors = {k: torch.as_tensor(data[k]) for k in ["mz", "intensity", "precursor", "adduct", "target_index"]}
    tensors["precursor"] = tensors["precursor"].float()
    fp = torch.as_tensor(data["fp_packed"]).to(device)
    shifts = torch.tensor([7, 6, 5, 4, 3, 2, 1, 0], device=device, dtype=torch.uint8)
    n = len(tensors["mz"])
    steps_total = epochs * max(n // batch, 1)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-4)
    scheduler = torch.optim.lr_scheduler.OneCycleLR(
        optimizer, max_lr=lr, total_steps=max(steps_total, 2), pct_start=0.05
    )
    loss_fn = nn.BCEWithLogitsLoss()
    use_amp = str(device).startswith("cuda")
    # Older GPUs such as the T4 have no native bfloat16, so they use float16 with loss scaling.
    amp_dtype = torch.bfloat16 if use_amp and torch.cuda.is_bf16_supported(including_emulation=False) else torch.float16
    scaler = torch.amp.GradScaler("cuda", enabled=use_amp and amp_dtype == torch.float16)
    autocast = (lambda: torch.autocast("cuda", dtype=amp_dtype)) if use_amp else contextlib.nullcontext
    xm = None
    if str(device).startswith("xla"):
        import torch_xla.core.xla_model as xm
    start, best, history = time.time(), -1.0, []
    q = val["queries"]
    q_adduct = adduct_ids(q.adduct)
    q_prec = q.precursor_mz.to_numpy()
    for epoch in range(epochs):
        model.train()
        order = torch.randperm(n)
        running, count = torch.zeros((), device=device), 0
        for i in range(0, n - batch + 1, batch):
            sel = order[i : i + batch]
            mz = tensors["mz"][sel].to(device)
            intensity = tensors["intensity"][sel].to(device)
            precursor = tensors["precursor"][sel].to(device)
            adduct = tensors["adduct"][sel].to(device)
            packed = fp[tensors["target_index"][sel].to(device)]
            target = ((packed.unsqueeze(-1) >> shifts) & 1).flatten(1).float()
            with autocast():
                loss = loss_fn(model(mz, intensity, precursor, adduct), target)
            optimizer.zero_grad(set_to_none=True)
            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            scaler.step(optimizer)
            scaler.update()
            if xm is not None:
                xm.mark_step()
            if scheduler.last_epoch < steps_total - 1:
                scheduler.step()
            running, count = running + loss.detach(), count + 1
            if (time.time() - start) / 60 > max_minutes:
                break
        probs = predict_probs(model, val["mz"], val["intensity"], q_prec, q_adduct, device)
        table = evaluate(
            rank_by_fingerprint(val["pool"], val["pool_fp"], q, probs), val["truth"], val["pool"].key.to_numpy()
        )
        score = float(table.loc["balanced", "MRR@25"])
        history.append(
            {
                "epoch": epoch + 1,
                "minutes": round((time.time() - start) / 60, 2),
                "train_bce": float(running.item()) / max(count, 1),
                "val_MRR@25": score,
                "class1": float(table.loc["class 1", "MRR@25"]),
                "class2": float(table.loc["class 2", "MRR@25"]),
                "np_proxy": float(table.loc["np proxy", "MRR@25"]),
            }
        )
        log(history[-1])
        if score > best:
            best = score
            torch.save({k: v.cpu() for k, v in model.state_dict().items()}, out_path)
        if (time.time() - start) / 60 > max_minutes:
            log("Stopped at the time budget.")
            break
    model.load_state_dict(torch.load(out_path, map_location="cpu"))
    return model, pd.DataFrame(history)


def fingerprint_metrics(spectra, probs, truth, pool, pool_fp):
    """Per-molecule fingerprint quality at a 0.5 threshold: Tanimoto and bit F1."""
    fp_index = {k: i for i, k in enumerate(pool.key.to_numpy())}
    spectra = spectra.copy()
    spectra["pos"] = np.arange(len(spectra))
    positions = spectra.groupby("molecule_id", sort=False).pos.apply(list)
    rows = []
    for row in truth.itertuples():
        predicted = probs[positions[row.molecule_id]].mean(axis=0) >= 0.5
        actual = np.unpackbits(pool_fp[fp_index[row.true_key]]).astype(bool)
        both, either = (predicted & actual).sum(), (predicted | actual).sum()
        denominator = predicted.sum() + actual.sum()
        rows.append((row.cls, both / either if either else 0.0, 2 * both / denominator if denominator else 0.0))
    res = pd.DataFrame(rows, columns=["cls", "tanimoto", "bit_f1"])
    return res.groupby("cls")[["tanimoto", "bit_f1"]].mean().round(4)


def get_device(prefer="auto"):
    """Pick CUDA, then TPU through torch_xla, then CPU."""
    import torch

    if prefer in ("auto", "cuda") and torch.cuda.is_available():
        return torch.device("cuda")
    if prefer in ("auto", "tpu"):
        try:
            import torch_xla

            return torch_xla.device()
        except Exception:
            pass
    print("WARNING: no GPU or TPU found, using CPU. Training will be very slow.")
    return torch.device("cpu")


def save_artifact(folder, model_name, model_kwargs, pool, pool_fp, settings):
    """Save everything a Kaggle submission needs next to model.pt, so RDKit isn't required there."""
    import json

    os.makedirs(folder, exist_ok=True)
    pool[["key", "smiles", "mass", "n_ref"]].to_parquet(os.path.join(folder, "pool.parquet"))
    np.save(os.path.join(folder, "pool_fp.npy"), pool_fp)
    with open(os.path.join(folder, "artifact.json"), "w") as handle:
        json.dump({"model_name": model_name, "model_kwargs": model_kwargs, "settings": settings}, handle, indent=2)


def load_artifact(folder, device):
    import json

    import torch

    with open(os.path.join(folder, "artifact.json")) as handle:
        info = json.load(handle)
    model = build_models()[info["model_name"]](**info["model_kwargs"])
    model.load_state_dict(torch.load(os.path.join(folder, "model.pt"), map_location="cpu"))
    pool = pd.read_parquet(os.path.join(folder, "pool.parquet"))
    return model.to(device), pool, np.load(os.path.join(folder, "pool_fp.npy")), info


# ---------- environment, run logs, figures ----------


def detect_platform():
    """Identify the runtime. Environment variables are checked first because a /kaggle
    folder or the google.colab package can exist outside their own platforms."""
    if os.environ.get("KAGGLE_KERNEL_RUN_TYPE"):
        return "kaggle"
    if os.environ.get("COLAB_RELEASE_TAG") or os.environ.get("COLAB_GPU") is not None:
        return "colab"
    try:
        import google.colab  # noqa: F401

        return "colab"
    except ImportError:
        pass
    return "kaggle" if os.path.exists("/kaggle/input") else "local"


def locate_on_drive(preferred, root="/content/drive/MyDrive", max_depth=4):
    """Return preferred if it holds the data, otherwise search Drive for a folder that does."""
    if os.path.exists(os.path.join(preferred, "train.parquet")):
        return preferred
    base_depth = root.rstrip(os.sep).count(os.sep)
    for folder, subfolders, files in os.walk(root):
        if "train.parquet" in files:
            print("Data found at", folder)
            return folder
        if folder.count(os.sep) - base_depth >= max_depth:
            subfolders[:] = []
    return preferred


def setup_environment(drive_data_dir, drive_run_root):
    """Return (platform, data_dir, run_root). On Colab, copy the data from Drive to fast local disk."""
    import shutil

    platform = detect_platform()
    if platform == "colab":
        from google.colab import drive

        drive.mount("/content/drive")
        drive_data_dir = locate_on_drive(drive_data_dir)
        local = "/content/casmi_data"
        os.makedirs(local, exist_ok=True)
        for name in ["train.parquet", "test.parquet", "sample_submission.csv"]:
            source, target = os.path.join(drive_data_dir, name), os.path.join(local, name)
            if not os.path.exists(source):
                raise FileNotFoundError(
                    f"{source} is missing. Upload the dataset folder to Drive or fix DRIVE_DATA_DIR."
                )
            if not os.path.exists(target) or os.path.getsize(target) != os.path.getsize(source):
                print("Copying", name, "from Drive")
                shutil.copy(source, target)
        return platform, local, drive_run_root
    if platform == "kaggle":
        return platform, find_data_dir(), "/kaggle/working/runs"
    return platform, find_data_dir(), "runs"


class Run:
    """One run folder: run.log, config.json, environment.json, metrics CSVs, figures, summary.json."""

    def __init__(self, name, root, config):
        import json
        import logging
        import platform as py_platform

        self.name = name
        self.root = root
        self.quick = bool(config.get("quick_run"))
        self.dir = os.path.join(root, f"{name}_{time.strftime('%Y%m%d_%H%M%S')}")
        self.fig_dir = os.path.join(self.dir, "figures")
        os.makedirs(self.fig_dir, exist_ok=True)
        self.start = time.time()
        self.logger = logging.getLogger(self.dir)
        self.logger.setLevel(logging.INFO)
        self.logger.propagate = False
        formatter = logging.Formatter("%(asctime)s | %(message)s", "%Y-%m-%d %H:%M:%S")
        for handler in (logging.FileHandler(os.path.join(self.dir, "run.log")), logging.StreamHandler(sys.stdout)):
            handler.setFormatter(formatter)
            self.logger.addHandler(handler)
        with open(os.path.join(self.dir, "config.json"), "w") as handle:
            json.dump(config, handle, indent=2, default=str)
        info = {
            "platform": detect_platform(),
            "python": py_platform.python_version(),
            "numpy": np.__version__,
            "pandas": pd.__version__,
        }
        for module in ["torch", "rdkit", "torch_xla"]:
            try:
                info[module] = __import__(module).__version__
            except Exception:
                info[module] = None
        if info["torch"]:
            import torch

            info["gpu"] = torch.cuda.get_device_name(0) if torch.cuda.is_available() else None
        with open(os.path.join(self.dir, "environment.json"), "w") as handle:
            json.dump(info, handle, indent=2)
        self.log(f"Run folder: {self.dir}")
        self.log(f"Environment: {info}")
        self.log(f"Config: {config}")

    def log(self, message):
        self.logger.info(str(message))

    def table(self, df, name, index=True):
        df.to_csv(os.path.join(self.dir, f"{name}.csv"), index=index)
        self.log(f"{name}:\n{df.to_string()}" if len(df) <= 50 else f"{name}: {len(df)} rows saved to {name}.csv")

    def figure(self, fig, name):
        import matplotlib.pyplot as plt

        path = os.path.join(self.fig_dir, f"{name}.png")
        fig.savefig(path, dpi=170, bbox_inches="tight", facecolor=fig.get_facecolor())
        plt.show()
        plt.close(fig)
        self.log(f"Saved figure {path}")

    def finish(self, entry):
        """Write summary.json and add this run to the shared experiment log."""
        import json

        entry = dict(
            entry, run_folder=os.path.basename(self.dir), runtime_minutes=round((time.time() - self.start) / 60, 2)
        )
        with open(os.path.join(self.dir, "summary.json"), "w") as handle:
            json.dump(entry, handle, indent=2, default=str)
        path = os.path.join(self.root, "experiment_log_quick.csv" if self.quick else "experiment_log.csv")
        log = pd.DataFrame([entry])
        if os.path.exists(path):
            log = pd.concat([pd.read_csv(path), log], ignore_index=True)
        log.to_csv(path, index=False)
        self.log(f"Experiment log updated: {path}")
        self.log(f"Summary: {entry}")
        return log


INK, MUTED, GRID, SURFACE = "#0b0b0b", "#52514e", "#e6e5e0", "#fcfcfb"
GROUP_COLORS = {"class 1": "#2a78d6", "class 2": "#eb6834", "np proxy": "#1baf7a"}
GROUP_LABELS = {
    "class 1": "Class 1 (has reference spectra)",
    "class 2": "Class 2 (no reference spectra)",
    "np proxy": "Natural-product proxy",
}


def _axes(ncols=1, size=(8.5, 4.6)):
    import matplotlib.pyplot as plt

    fig, axes = plt.subplots(1, ncols, figsize=size, facecolor=SURFACE, squeeze=False)
    for ax in axes[0]:
        ax.set_facecolor(SURFACE)
        for side in ("top", "right"):
            ax.spines[side].set_visible(False)
        for side in ("left", "bottom"):
            ax.spines[side].set_color(GRID)
        ax.tick_params(colors=MUTED, length=0)
        ax.yaxis.grid(True, color=GRID, linewidth=0.8)
        ax.set_axisbelow(True)
    return fig, axes[0]


def _title(ax, text):
    ax.set_title(text, loc="left", color=INK, fontsize=12, pad=12)


def _labels(ax, x, y):
    ax.set_xlabel(x, color=MUTED)
    ax.set_ylabel(y, color=MUTED)


def plot_group_bars(frame, title, x_label=""):
    """Grouped MRR@25 bars. frame: index = configuration, columns = class 1, class 2, np proxy."""
    fig, (ax,) = _axes(size=(max(8.5, 1.5 * len(frame) + 3), 4.8))
    width = 0.8 / 3
    positions = np.arange(len(frame))
    for g, group in enumerate(GROUP_COLORS):
        values = frame[group].astype(float).to_numpy()
        bars = ax.bar(
            positions + (g - 1) * width, values, width * 0.92, color=GROUP_COLORS[group], label=GROUP_LABELS[group]
        )
        for bar, value in zip(bars, values, strict=False):
            ax.text(
                bar.get_x() + bar.get_width() / 2, value + 0.012, f"{value:.2f}", ha="center", color=MUTED, fontsize=7.5
            )
    ax.set_xticks(positions, frame.index, color=MUTED, fontsize=9)
    ax.set_ylim(0, 1.08)
    _labels(ax, x_label, "Validation MRR@25")
    _title(ax, title)
    ax.legend(frameon=False, labelcolor=MUTED, ncols=3, loc="upper center", bbox_to_anchor=(0.5, -0.16))
    return fig


def plot_hit_curve(ranks, title):
    """Share of molecules whose true structure is within the top k."""
    fig, (ax,) = _axes()
    ks = np.arange(1, TOP_K + 1)
    parts = {"class 1": ranks[ranks.cls == 1], "class 2": ranks[ranks.cls == 2], "np proxy": ranks[ranks.np_proxy]}
    ends = []
    for group, part in parts.items():
        if len(part) == 0:
            continue
        values = [part["rank"].between(1, k).mean() for k in ks]
        ax.plot(ks, values, color=GROUP_COLORS[group], linewidth=2, label=GROUP_LABELS[group])
        ends.append([values[-1], group])
    ends.sort()
    for i in range(1, len(ends)):  # keep the end labels from overlapping
        ends[i][0] = max(ends[i][0], ends[i - 1][0] + 0.055)
    for value, group in ends:
        ax.text(TOP_K + 0.3, value, group, color=MUTED, fontsize=9, va="center")
    ax.set_xlim(1, TOP_K + 3.5)
    ax.set_ylim(0, 1.08)
    _labels(ax, "k (list position)", "Share found in top k")
    _title(ax, title)
    ax.legend(frameon=False, labelcolor=MUTED, ncols=3, loc="upper center", bbox_to_anchor=(0.5, -0.16))
    return fig


def plot_rank_histogram(ranks, title):
    """Where the true structure landed. The last bar counts misses."""
    fig, (ax,) = _axes()
    counts = [(ranks["rank"] == r).sum() for r in range(1, TOP_K + 1)] + [(ranks["rank"] == 0).sum()]
    ax.bar(np.arange(1, TOP_K + 2), counts, 0.8, color=GROUP_COLORS["class 1"])
    ax.set_xticks([1, 5, 10, 15, 20, 25, TOP_K + 1], ["1", "5", "10", "15", "20", "25", "missed"], color=MUTED)
    _labels(ax, "Rank of true structure", "Molecules")
    _title(ax, title)
    return fig


def plot_histogram(values, title, x_label, marker=None, marker_label=None, bins=50):
    fig, (ax,) = _axes()
    ax.hist(values, bins=bins, color=GROUP_COLORS["class 1"], rwidth=0.92)
    if marker is not None:
        ax.axvline(marker, color=INK, linewidth=1.2, linestyle="--")
        ax.text(marker, ax.get_ylim()[1] * 0.95, f" {marker_label}", color=MUTED, fontsize=9)
    _labels(ax, x_label, "Molecules")
    _title(ax, title)
    return fig


def plot_training(history, title):
    """Loss and validation MRR in separate panels, because they have different scales."""
    fig, (left, right) = _axes(2, (11, 4.4))
    left.plot(history.epoch, history.train_bce, color=GROUP_COLORS["class 1"], linewidth=2, marker="o", markersize=5)
    _labels(left, "Epoch", "Training BCE")
    _title(left, "Training loss")
    for column, group in [("class1", "class 1"), ("class2", "class 2"), ("np_proxy", "np proxy")]:
        right.plot(
            history.epoch,
            history[column].astype(float),
            color=GROUP_COLORS[group],
            linewidth=2,
            marker="o",
            markersize=5,
            label=GROUP_LABELS[group],
        )
    _labels(right, "Epoch", "Validation MRR@25")
    _title(right, "Validation by class")
    right.legend(frameon=False, labelcolor=MUTED, loc="upper center", bbox_to_anchor=(0.5, -0.18), ncols=1)
    from matplotlib.ticker import MaxNLocator

    for ax in (left, right):
        ax.xaxis.set_major_locator(MaxNLocator(integer=True))
    fig.suptitle(title, x=0.06, ha="left", color=INK, fontsize=13)
    fig.tight_layout()
    return fig


def plot_comparison(log, title="Model comparison on validation"):
    """Latest run of each version from the experiment log."""
    latest = log.drop_duplicates("version", keep="last").set_index("version")
    frame = latest.rename(columns={"class1": "class 1", "class2": "class 2", "np_proxy": "np proxy"})[
        list(GROUP_COLORS)
    ]
    return plot_group_bars(frame, title)
```

```python
%%writefile m7/common/pooling.py
"""Spectrum encoder and the three ways to pool a molecule's spectra. Owner: Member 7.

Every query (molecule) has 1-16 spectra. Each spectrum is encoded on its own into an
embedding ``h`` and a per-spectrum fingerprint prediction. The pooling heads turn the
variable-size set {h_1..h_n} into ONE fingerprint prediction for the molecule:

- ``mean_probabilities``: no training, average the per-spectrum fingerprint probabilities
  (what Member 2's CNN scorer does; our baseline to beat).
- ``DeepSetsPool`` (arch2): rho(mean_i phi(h_i)). Order-free, any set size, every
  spectrum counts equally.
- ``AttentionPool`` (arch1): gated attention (Ilse et al., 2018). Learns a weight per
  spectrum, so noisy or uninformative spectra can be down-weighted.

Sets are padded to the largest set in a batch; ``mask`` is True for real spectra.
"""

from __future__ import annotations

import torch
from torch import nn


def mlp(sizes: list[int], dropout: float = 0.0, last_act: bool = True) -> nn.Sequential:
    layers: list[nn.Module] = []
    for i, (a, b) in enumerate(zip(sizes[:-1], sizes[1:], strict=True)):
        layers.append(nn.Linear(a, b))
        if i < len(sizes) - 2 or last_act:
            layers += [nn.ReLU(), nn.Dropout(dropout)]
    return nn.Sequential(*layers)


class SpectrumEncoder(nn.Module):
    """Binned spectrum (+ metadata) -> embedding h, plus a per-spectrum fingerprint head."""

    def __init__(self, n_bins: int, n_meta: int, fp_size: int, hidden: int = 1024, emb: int = 512,
                 dropout: float = 0.2):
        super().__init__()
        self.body = mlp([n_bins + n_meta, hidden, emb], dropout)
        self.head = nn.Linear(emb, fp_size)

    def embed(self, x: torch.Tensor, meta: torch.Tensor) -> torch.Tensor:
        return self.body(torch.cat([x, meta], dim=-1))

    def forward(self, x: torch.Tensor, meta: torch.Tensor) -> torch.Tensor:
        return self.head(self.embed(x, meta))


def mean_probabilities(probs: torch.Tensor, mask: torch.Tensor) -> torch.Tensor:
    """Average per-spectrum probabilities over the real spectra of each set.

    probs: (B, N, F), mask: (B, N) bool -> (B, F)
    """
    m = mask.unsqueeze(-1).to(probs.dtype)
    return (probs * m).sum(1) / m.sum(1).clamp(min=1)


class DeepSetsPool(nn.Module):
    """rho( mean_i phi(h_i) ), with log(set size) appended so rho knows how much evidence it has."""

    def __init__(self, emb: int, fp_size: int, hidden: int = 512, dropout: float = 0.2):
        super().__init__()
        self.phi = mlp([emb, hidden, hidden], dropout)
        self.rho = mlp([hidden + 1, hidden, fp_size], dropout, last_act=False)

    def forward(self, h: torch.Tensor, mask: torch.Tensor):
        m = mask.unsqueeze(-1).to(h.dtype)
        n = m.sum(1).clamp(min=1)
        pooled = (self.phi(h) * m).sum(1) / n
        return self.rho(torch.cat([pooled, torch.log(n)], dim=-1)), None


class AttentionPool(nn.Module):
    """Gated attention pooling: weights a_i = softmax_i( w . (tanh(V h_i) * sigmoid(U h_i)) ).

    Returns logits (B, F) and the attention weights (B, N), which sum to 1 over real spectra.
    """

    def __init__(self, emb: int, fp_size: int, hidden: int = 512, att: int = 128, dropout: float = 0.2):
        super().__init__()
        self.phi = mlp([emb, hidden, hidden], dropout)
        self.V = nn.Linear(hidden, att)
        self.U = nn.Linear(hidden, att)
        self.w = nn.Linear(att, 1)
        self.rho = mlp([hidden + 1, hidden, fp_size], dropout, last_act=False)

    def forward(self, h: torch.Tensor, mask: torch.Tensor):
        z = self.phi(h)
        scores = self.w(torch.tanh(self.V(z)) * torch.sigmoid(self.U(z))).squeeze(-1)
        scores = scores.masked_fill(~mask, float("-inf"))
        weights = torch.softmax(scores, dim=1)
        pooled = (weights.unsqueeze(-1) * z).sum(1)
        n = mask.sum(1, keepdim=True).to(h.dtype).clamp(min=1)
        return self.rho(torch.cat([pooled, torch.log(n)], dim=-1)), weights


def build_pool(name: str, emb: int, fp_size: int, **kw) -> nn.Module:
    pools = {"deepsets": DeepSetsPool, "attention": AttentionPool}
    if name not in pools:
        raise ValueError(f"Unknown pooling {name!r}; choose from {sorted(pools)}")
    return pools[name](emb, fp_size, **kw)
```

```python
%%writefile m7/common/sets.py
"""Turn rows of spectra into per-molecule sets, and pad sets into batches. Owner: Member 7.

Training molecules can have hundreds of spectra across libraries, while a test molecule has
1-16 (median 3). To train pooling on the situation it will face, each training step draws a
random subset whose size follows that test-like distribution.
"""

from __future__ import annotations

import numpy as np
import torch

MAX_SET = 16  # hidden test: 1-16 spectra per molecule, median 3


def draw_set_sizes(rng: np.random.Generator, n: int, max_set: int = MAX_SET) -> np.ndarray:
    """Draw set sizes in [1, max_set] with median 3 (geometric, p=0.25, capped)."""
    return np.minimum(rng.geometric(0.25, size=n), max_set)


def group_indices(keys: np.ndarray) -> dict[str, np.ndarray]:
    """Map each molecule key to the row indices of its spectra (order preserved)."""
    order = np.argsort(keys, kind="stable")
    sorted_keys = keys[order]
    starts = np.flatnonzero(np.r_[True, sorted_keys[1:] != sorted_keys[:-1]])
    ends = np.r_[starts[1:], len(order)]
    return {sorted_keys[s]: order[s:e] for s, e in zip(starts, ends, strict=True)}


def sample_sets(groups: dict[str, np.ndarray], keys: list[str], rng: np.random.Generator,
                max_set: int = MAX_SET) -> list[np.ndarray]:
    """For each key, sample a test-like number of its spectra without replacement."""
    sizes = draw_set_sizes(rng, len(keys), max_set)
    out = []
    for key, size in zip(keys, sizes, strict=True):
        rows = groups[key]
        out.append(rng.choice(rows, size=min(int(size), len(rows)), replace=False))
    return out


def pad_sets(sets: list[np.ndarray], features: np.ndarray | torch.Tensor) -> tuple[torch.Tensor, torch.Tensor]:
    """Gather per-spectrum features into (B, N_max, D) with a (B, N_max) mask of real rows.

    Output stays on the device of ``features`` (CPU array or GPU tensor).
    """
    feats = torch.as_tensor(features)
    n_max = max(len(s) for s in sets)
    index = np.zeros((len(sets), n_max), dtype=np.int64)
    mask = np.zeros((len(sets), n_max), dtype=bool)
    for i, rows in enumerate(sets):
        index[i, : len(rows)] = rows
        mask[i, : len(rows)] = True
    index_t = torch.from_numpy(index).to(feats.device)
    mask_t = torch.from_numpy(mask).to(feats.device)
    out = feats[index_t] * mask_t.unsqueeze(-1).to(feats.dtype)
    return out, mask_t
```

```python
%%writefile m7/common/full.py
"""Full pooling run on Member 3's pipeline: embeddings, pooled ranking, fair control, export.

Member 3's ``casmi_pipeline`` (``baselines/b1_mass_only``) provides the validation split
(classes 1 and 2), the candidate pool, the per-spectrum fingerprint model and the
evaluation. This module adds what pooling needs on top of it. Owner: Member 7.

Pooling methods compared on the same frozen per-spectrum model:

- ``mean``: average of per-spectrum probabilities (what ``rank_by_fingerprint`` does).
- ``mean_tuned``: fair control. The per-spectrum head gets the same extra training as the
  poolers, with the loss on the averaged probabilities, but no spectrum sees another.
  If the poolers only beat ``mean`` and not ``mean_tuned``, the gain is extra training,
  not pooling.
- ``deepsets`` (arch2) and ``attention`` (arch1).
"""

from __future__ import annotations

import copy
import json
import os

import numpy as np
import pandas as pd
import torch
import torch.nn.functional as F

from .pooling import build_pool
from .sets import pad_sets, sample_sets

METHODS = ["mean", "mean_tuned", "deepsets", "attention"]
KAGGLE_METHODS = ["mean", "mean_tuned", "attention_head"]


# ---------- per-spectrum model (Member 3's ResidualMLP) ----------


def embed_batch(model, mz, intensity, precursor, adduct):
    """Hidden vector h of ResidualMLP, i.e. everything before ``model.head``."""
    bins = mz.long().clamp(0, model.n_bins - 1)
    x = torch.zeros(mz.size(0), model.n_bins, device=mz.device, dtype=intensity.dtype)
    x.scatter_add_(1, bins, intensity)
    x = torch.cat([x, (precursor / 1000.0).unsqueeze(1), model.adduct(adduct)], dim=1)
    return model.blocks(model.stem(x))


@torch.no_grad()
def embed_all(model, mz, intensity, precursor, adduct, device, batch=2048, dtype=torch.float16):
    """Embeddings for many spectra, returned on the CPU (they can be millions of rows)."""
    model.eval()
    out = []
    for i in range(0, len(mz), batch):
        sl = slice(i, i + batch)
        h = embed_batch(model,
                        torch.as_tensor(mz[sl], device=device),
                        torch.as_tensor(intensity[sl], device=device),
                        torch.as_tensor(precursor[sl], dtype=torch.float32, device=device),
                        torch.as_tensor(adduct[sl], device=device))
        out.append(h.to(dtype).cpu())
    return torch.cat(out)


def unpack_bits(packed: torch.Tensor) -> torch.Tensor:
    """(B, F/8) uint8 -> (B, F) float, MSB first like np.packbits."""
    shifts = torch.arange(7, -1, -1, device=packed.device, dtype=torch.uint8)
    return ((packed.unsqueeze(-1) >> shifts) & 1).flatten(1).float()


# ---------- stage B: pooling heads and the fair control ----------


class TunedMean(torch.nn.Module):
    """Control: per-spectrum head, then the average of probabilities. Same call signature as the pools."""

    def __init__(self, head: torch.nn.Module):
        super().__init__()
        self.head = copy.deepcopy(head)
        for p in self.head.parameters():  # the encoder is frozen; this copy is trained
            p.requires_grad_(True)

    def forward(self, h, mask):
        probs = torch.sigmoid(self.head(h).float())
        m = mask.unsqueeze(-1).to(probs.dtype)
        mean = (probs * m).sum(1) / m.sum(1).clamp(min=1)
        return torch.logit(mean.clamp(1e-6, 1 - 1e-6)), None


class AttentionHead(torch.nn.Module):
    """Attention that keeps the trained per-spectrum head.

    The full-run notebook showed DeepSets/attention lose because they replace the trained
    head with a fresh one. Here attention only decides how much each spectrum's
    probabilities count; the head itself is a trainable copy, exactly as in ``TunedMean``.
    With uniform weights this is ``TunedMean``.
    """

    def __init__(self, head: torch.nn.Module, emb: int, att: int = 128):
        super().__init__()
        self.head = copy.deepcopy(head)
        for p in self.head.parameters():
            p.requires_grad_(True)
        self.V = torch.nn.Linear(emb, att)
        self.U = torch.nn.Linear(emb, att)
        self.w = torch.nn.Linear(att, 1)
        torch.nn.init.zeros_(self.w.weight)  # start as the plain average
        torch.nn.init.zeros_(self.w.bias)

    def forward(self, h, mask):
        scores = self.w(torch.tanh(self.V(h)) * torch.sigmoid(self.U(h))).squeeze(-1)
        weights = torch.softmax(scores.float().masked_fill(~mask, float("-inf")), dim=1)
        probs = torch.sigmoid(self.head(h).float())
        mean = (weights.unsqueeze(-1) * probs).sum(1)
        return torch.logit(mean.clamp(1e-6, 1 - 1e-6)), weights


def make_pool(name, emb, fp_bits, head=None):
    if name == "mean_tuned":
        return TunedMean(head)
    if name == "attention_head":
        return AttentionHead(head, emb)
    return build_pool(name, emb, fp_bits)


def batched_forward(pool, sets, H, device, chunk=1024):
    """Run a pool over many sets. H lives on the CPU; each chunk is gathered then moved."""
    logits, weights = [], []
    for s in range(0, len(sets), chunk):
        h, mask = pad_sets(sets[s:s + chunk], H)
        lg, w = pool(h.to(device).float(), mask.to(device))
        logits.append(lg.float())
        if w is not None:
            weights += [w[i, :len(sets[s + i])].detach().cpu().numpy() for i in range(len(w))]
    return torch.cat(logits), (weights or None)


def train_pool(pool, H, groups, train_keys, target_fp, device, H_val, val_sets, val_bits,
               epochs=20, batch=256, lr=1e-3, seed=0, log=print, ckpt=None, score_fn=None):
    """Train one pooling head on test-like random subsets of each training molecule's spectra.

    H: train embeddings (CPU). target_fp: dict key -> packed fingerprint (uint8).
    H_val, val_sets, val_bits: fixed validation sets and their (n_sets, F) 0/1 targets.
    Keeps the epoch with the lowest validation BCE, or, with score_fn (validation probabilities ->
    MRR@25, higher is better), the epoch with the best ranking. Lower BCE does not imply better ranking.
    """
    torch.manual_seed(seed)
    rng = np.random.default_rng(seed)
    pool = pool.to(device)
    opt = torch.optim.AdamW(pool.parameters(), lr=lr, weight_decay=1e-4)
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=max(epochs, 1))
    best, best_state, hist = float("inf"), None, []
    for ep in range(epochs):
        pool.train()
        order = rng.permutation(len(train_keys))
        total, n = 0.0, 0
        for s in range(0, len(order), batch):
            keys = [train_keys[j] for j in order[s:s + batch]]
            h, mask = pad_sets(sample_sets(groups, keys, rng), H)
            logits, _ = pool(h.to(device).float(), mask.to(device))
            target = unpack_bits(torch.as_tensor(np.stack([target_fp[k] for k in keys]), device=device))
            loss = F.binary_cross_entropy_with_logits(logits, target)
            opt.zero_grad(set_to_none=True)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(pool.parameters(), 1.0)
            opt.step()
            total, n = total + loss.item(), n + 1
        sched.step()
        pool.eval()
        with torch.no_grad():
            logits, _ = batched_forward(pool, val_sets, H_val, device)
            vl = float(F.binary_cross_entropy_with_logits(logits, val_bits.to(device)))
        row = {"epoch": ep + 1, "train_bce": total / max(n, 1), "val_bce": vl}
        crit = vl
        if score_fn is not None:
            row["val_mrr"] = float(score_fn(torch.sigmoid(logits.float()).cpu().numpy()))
            crit = -row["val_mrr"]
        hist.append(row)
        log(f"  epoch {ep + 1}: train {row['train_bce']:.4f}  val {vl:.4f}  mrr {row.get('val_mrr', float('nan')):.4f}")
        if crit < best:
            best, best_state = vl, {k: v.detach().cpu().clone() for k, v in pool.state_dict().items()}
    pool.load_state_dict(best_state)
    if ckpt:
        torch.save(best_state, ckpt)
    return pool.eval(), pd.DataFrame(hist)


# ---------- ranking with Member 3's pool and evaluation ----------


def window_similarity(cp, pool, pool_fp_packed, masses, probs, ppm=10.0):
    """{molecule: (window indices, soft Tanimoto)}: the expensive part of ranking, computed once."""
    pool_mass = pool.mass.to_numpy()
    out = {}
    for (mol, mass), p in zip(masses.items(), probs, strict=True):
        idx = cp.window_indices(pool_mass, mass, ppm)
        tanimoto = np.zeros(0, dtype=np.float32)
        if len(idx):
            bits = np.unpackbits(pool_fp_packed[idx], axis=1).astype(np.float32)
            overlap = bits @ p
            tanimoto = overlap / (p.sum() + bits.sum(axis=1) - overlap + 1e-8)
        out[mol] = (idx, tanimoto)
    return out


def rank_from_similarity(cp, pool, sims, masses, k=25, prior=None, extra=None):
    """Rank each window by similarity + optional per-structure prior + optional per-molecule extra term."""
    pool_mass = pool.mass.to_numpy()
    rankings = {}
    for mol, mass in masses.items():
        idx, score = sims[mol]
        if len(idx):
            if prior is not None:  # per-structure bonus, e.g. for structures in a curated database
                score = score + prior[idx]
            if extra is not None:  # per-molecule term over the window, e.g. fragment-formula explanation
                score = score + extra(mol, idx)
            idx = idx[np.argsort(-score, kind="stable")]
        rankings[mol] = cp.fill_to_k(idx, pool_mass, mass, k=k)
    return rankings


def rank_pooled(cp, pool, pool_fp_packed, masses, probs, ppm=10.0, k=25, prior=None, extra=None):
    """Rank candidates for each molecule by soft Tanimoto to ONE pooled probability vector.

    Same as ``cp.rank_by_fingerprint`` except the pooling has already happened.
    masses: Series molecule_id -> neutral mass. probs: (n_molecules, F) in the same order.
    """
    sims = window_similarity(cp, pool, pool_fp_packed, masses, probs, ppm)
    return rank_from_similarity(cp, pool, sims, masses, k=k, prior=prior, extra=extra)


def rrf_fuse(rank_lists, weights=None, k=60, top=25):
    """Reciprocal-rank fusion of several ranked lists of pool indices (best first)."""
    weights = weights or [1.0] * len(rank_lists)
    score = {}
    for lst, w in zip(rank_lists, weights, strict=True):
        for r, idx in enumerate(lst):
            score[int(idx)] = score.get(int(idx), 0.0) + w / (k + r + 1)
    return sorted(score, key=lambda i: (-score[i], i))[:top]


def paired_bootstrap(rr_a, rr_b, n=2000, seed=3):
    """Mean difference a - b in reciprocal rank with a 95% CI (same molecules, resampled together)."""
    diff = np.asarray(rr_a, float) - np.asarray(rr_b, float)
    if len(diff) == 0:
        return {"delta": np.nan, "ci_low": np.nan, "ci_high": np.nan}
    draws = np.random.default_rng(seed).choice(diff, size=(n, len(diff))).mean(axis=1)
    return {"delta": float(diff.mean()), "ci_low": float(np.quantile(draws, 0.025)),
            "ci_high": float(np.quantile(draws, 0.975))}


# ---------- export for the offline Kaggle notebook (Member 6) ----------


def save_artifact(folder, encoder_kwargs, encoder_state, method, pool_state, emb, fp_bits, settings):
    """Everything needed to rank test molecules offline, except the candidate pool files."""
    os.makedirs(folder, exist_ok=True)
    torch.save(encoder_state, os.path.join(folder, "encoder.pt"))
    torch.save(pool_state, os.path.join(folder, f"pool_{method}.pt"))
    info = {"encoder": "residual_mlp", "encoder_kwargs": encoder_kwargs, "method": method,
            "emb": emb, "fp_bits": fp_bits, "settings": settings}
    with open(os.path.join(folder, "pooling_artifact.json"), "w") as f:
        json.dump(info, f, indent=2)


def load_artifact(folder, cp, device, head_for_control=None):
    with open(os.path.join(folder, "pooling_artifact.json")) as f:
        info = json.load(f)
    enc = cp.build_models()[info["encoder"]](**info["encoder_kwargs"])
    enc.load_state_dict(torch.load(os.path.join(folder, "encoder.pt"), map_location="cpu"))
    enc = enc.to(device).eval()
    pool = make_pool(info["method"], info["emb"], info["fp_bits"], head=enc.head)
    pool.load_state_dict(torch.load(os.path.join(folder, f"pool_{info['method']}.pt"), map_location="cpu"))
    return enc, pool.to(device).eval(), info


@torch.no_grad()
def predict_molecules(cp, enc, pool, spectra, mz, intensity, device):
    """Pooled fingerprint probabilities per molecule.

    spectra: rows with molecule_id, precursor_mz, adduct, neutral_mass, aligned with mz/intensity.
    Returns (masses Series molecule_id -> median neutral mass, probs array in the same order).
    """
    H = embed_all(enc, mz, intensity, spectra.precursor_mz.to_numpy(), cp.adduct_ids(spectra.adduct), device)
    pos = pd.Series(np.arange(len(spectra)), index=spectra.index)
    groups = spectra.assign(pos=pos.to_numpy()).groupby("molecule_id", sort=False)
    sets = [g.pos.to_numpy() for _, g in groups]
    masses = groups.neutral_mass.median()
    if isinstance(pool, str) and pool == "mean":
        probs = []
        for s in range(0, len(sets), 1024):
            h, mask = pad_sets(sets[s:s + 1024], H)
            p = torch.sigmoid(enc.head(h.to(device).float()).float())
            m = mask.to(device).unsqueeze(-1).float()
            probs.append(((p * m).sum(1) / m.sum(1).clamp(min=1)).cpu())
        return masses, torch.cat(probs).numpy()
    logits, _ = batched_forward(pool, sets, H, device)
    return masses, torch.sigmoid(logits).cpu().numpy()
```

```python
%%writefile m7/common/candidates.py
"""External candidates (COCONUT, PubChemLite) and gated B2 fusion for the offline Kaggle notebook.

Candidates: every attached file whose path contains a source name is read, its SMILES column is parsed
with RDKit (largest fragment kept), and each structure becomes (inchikey14, smiles, exact mass, Morgan
fingerprint). Parsing runs in worker processes; ~1M structures take a few minutes on Kaggle's 4 CPUs.

Gate: B2 (library search) is excellent when the structure has library spectra and useless otherwise.
Use the B2 ranking only for molecules whose best library match scores at least tau, else the model's.
tau is chosen on validation; the honest estimate is the 2-fold cross-fitted score.
"""

from __future__ import annotations

import glob
import os
import time
import zlib
from multiprocessing import get_context

import numpy as np
import pandas as pd

SOURCES = ("coconut", "pubchemlite")
SMILES_COLUMNS = ("canonical_smiles", "smiles", "isomeric_smiles", "normalized_smiles")
EXTENSIONS = (".csv", ".csv.gz", ".zip", ".parquet", ".tsv")
MASS_RANGE = (100.0, 1500.0)  # hidden test spans 157-1,159 Da
MAX_SMILES_LEN = 400  # longer SMILES are polymers/salts/annotations, far outside the test chemistry


# ---------- external candidates ----------


def find_files(root="/kaggle/input"):
    """{source: [files]} for attached files whose path names a source."""
    found = {}
    for f in sorted(glob.glob(os.path.join(root, "**", "*"), recursive=True)):
        low = f.lower()
        if os.path.isfile(f) and low.endswith(EXTENSIONS):
            for name in SOURCES:
                if name in low:
                    found.setdefault(name, []).append(f)
    return found


def smiles_column(columns):
    low = {c.lower(): c for c in columns}
    for c in SMILES_COLUMNS:
        if c in low:
            return low[c]
    for c in columns:
        if "smiles" in c.lower():
            return c
    raise ValueError(f"no SMILES column in {list(columns)[:20]}")


def read_smiles(path):
    """Unique SMILES strings from a csv / tsv / zip / parquet file."""
    if path.lower().endswith(".parquet"):
        df = pd.read_parquet(path)
        return df[smiles_column(df.columns)].dropna().astype(str).unique()
    sep = "\t" if ".tsv" in path.lower() else ","
    header = pd.read_csv(path, sep=sep, nrows=0)
    if len(header.columns) == 1 and "\t" in header.columns[0]:
        sep = "\t"
        header = pd.read_csv(path, sep=sep, nrows=0)
    col = smiles_column(header.columns)
    df = pd.read_csv(path, sep=sep, usecols=[col], dtype=str, on_bad_lines="skip")
    return df[col].dropna().astype(str).unique()


def _convert(args):
    """Worker: SMILES chunk -> keys, canonical smiles, masses, formulas, packed fingerprints."""
    chunk, n_bits = args
    from rdkit import Chem, RDLogger
    from rdkit.Chem import Descriptors, rdFingerprintGenerator
    from rdkit.Chem.MolStandardize import rdMolStandardize
    from rdkit.Chem.rdMolDescriptors import CalcMolFormula

    RDLogger.DisableLog("rdApp.*")
    gen = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=n_bits)
    chooser = rdMolStandardize.LargestFragmentChooser()
    keys, smis, masses, formulas, fps, kept = [], [], [], [], [], []
    for pos, s in enumerate(chunk):
        try:
            mol = Chem.MolFromSmiles(s)
            if mol is None:
                continue
            if len(Chem.GetMolFrags(mol)) > 1:
                mol = chooser.choose(mol)
            mass = Descriptors.ExactMolWt(mol)
            if not MASS_RANGE[0] <= mass <= MASS_RANGE[1]:
                continue
            key = Chem.MolToInchiKey(mol)[:14]
            if len(key) != 14:
                continue
            keys.append(key)
            smis.append(Chem.MolToSmiles(mol))
            masses.append(mass)
            formulas.append(CalcMolFormula(mol))
            fps.append(np.packbits(gen.GetFingerprintAsNumPy(mol).astype(bool)))
            kept.append(pos)
        except Exception:  # noqa: BLE001 - one bad record must not stop the run
            continue
    fp = np.stack(fps) if fps else np.zeros((0, n_bits // 8), dtype=np.uint8)
    return keys, smis, masses, formulas, fp, kept


def load_candidates(files, n_bits=2048, workers=None, chunk=20000, log=print, smiles_by_source=None):
    """files: {source: [paths]}; smiles_by_source: {source: SMILES array} already read (e.g. PubChem windows).

    Returns (frame[key, smiles, mass, source], packed fp) with unique keys; earlier sources win duplicates.
    """
    frames, fps = [], []
    workers = workers or os.cpu_count() or 1
    sources = {s: (lambda paths=paths: np.concatenate([read_smiles(p) for p in paths])) for s, paths in files.items()}
    for s, arr in (smiles_by_source or {}).items():
        sources[s] = (lambda arr=arr: np.asarray(arr, dtype=object))
    for source, get in sources.items():
        t0 = time.time()
        # pandas keeps Python strings; numpy .astype(str) would pad every SMILES to the longest one (OOM)
        smiles = pd.Series(get(), dtype=object).dropna().astype(str)
        smiles = smiles[smiles.str.len() <= MAX_SMILES_LEN].unique()
        jobs = [(smiles[i:i + chunk].tolist(), n_bits) for i in range(0, len(smiles), chunk)]
        with get_context("fork").Pool(workers) as pool:
            parts = pool.map(_convert, jobs)
        keys = [k for p in parts for k in p[0]]
        frame = pd.DataFrame({"key": keys, "smiles": [s for p in parts for s in p[1]],
                              "mass": [m for p in parts for m in p[2]], "formula": [f for p in parts for f in p[3]],
                              "source": source})
        frames.append(frame)
        fps.append(np.concatenate([p[4] for p in parts]))
        log(f"{source}: {len(smiles):,} SMILES -> {len(frame):,} structures in range "
            f"({frame.key.nunique():,} unique) in {time.time() - t0:.0f}s")
    if not frames:
        return None, None
    frame = pd.concat(frames, ignore_index=True)
    fp = np.concatenate(fps)
    keep = ~frame.key.duplicated().to_numpy()
    return frame[keep].reset_index(drop=True), fp[keep]


# ---------- PubChem, restricted to the query mass windows ----------


def find_pubchem(root="/kaggle/input"):
    """Paths of PubChem's CID-Mass and CID-SMILES files (gzipped or not), or None."""
    found = {}
    for f in glob.glob(os.path.join(root, "**", "*"), recursive=True):
        name = os.path.basename(f).lower()
        if os.path.isfile(f) and name.startswith("cid-mass"):
            found["mass"] = f
        elif os.path.isfile(f) and name.startswith("cid-smiles"):
            found["smiles"] = f
    return found if len(found) == 2 else None


def _compression(path):
    with open(path, "rb") as f:
        return "gzip" if f.read(2) == b"\x1f\x8b" else None


ORGANIC = r"^(?:(?:C|H|N|O|P|S|F|Cl|Br|I)\d*)+$"  # neutral formulas made only of these elements


def pubchem_in_windows(mass_path, smiles_path, masses, ppm=10.0, cap=None, chunk=5_000_000, log=print):
    """SMILES of PubChem compounds within ppm of a query mass, at most `cap` per query.

    Keeps neutral, organic-element formulas. PubChem is dense (~8,000 compounds per typical 10 ppm window);
    cap=None keeps them all, an integer keeps the `cap` lowest CIDs (longest-known) per window.
    Streams both files once; adapts to whatever test set is attached.
    """
    q = np.unique(np.asarray(masses, dtype=float))
    q = q[np.isfinite(q)]
    if len(q) == 0:
        return np.array([], dtype=object)
    q = np.r_[q, q[-1]] if len(q) == 1 else q
    t0, cids, monos, n = time.time(), [], [], 0
    for df in pd.read_csv(mass_path, sep="\t", header=None, names=["cid", "formula", "mono", "exact"],
                          usecols=["cid", "formula", "mono"],
                          dtype={"cid": np.int64, "formula": str, "mono": np.float64},
                          chunksize=chunk, compression=_compression(mass_path), on_bad_lines="skip"):
        m = df.mono.to_numpy()
        i = np.searchsorted(q, m).clip(1, len(q) - 1)
        near = np.minimum(np.abs(m - q[i]), np.abs(m - q[i - 1])) <= m * ppm * 1e-6
        sub = df[near]
        sub = sub[sub.formula.str.fullmatch(ORGANIC, na=False)]
        cids.append(sub.cid.to_numpy())
        monos.append(sub.mono.to_numpy())
        n += len(df)
    cid, mono = np.concatenate(cids), np.concatenate(monos)
    order = np.argsort(mono, kind="stable")
    cid, mono = cid[order], mono[order]
    keep, truncated = [], 0
    for x in q:
        tol = x * ppm * 1e-6
        lo, hi = np.searchsorted(mono, x - tol), np.searchsorted(mono, x + tol, side="right")
        window = cid[lo:hi]
        if cap and len(window) > cap:
            truncated += 1
            window = np.partition(window, cap - 1)[:cap]
        keep.append(window)
    keep = np.unique(np.concatenate(keep))
    log(f"pubchem: {n:,} scanned, {len(cid):,} organic & neutral in windows; kept {len(keep):,} "
        f"(cap {cap}: {truncated} of {len(q)} windows truncated) ({time.time() - t0:.0f}s)")
    out = []
    for df in pd.read_csv(smiles_path, sep="\t", header=None, names=["cid", "smiles"],
                          dtype={"cid": np.int64, "smiles": str}, chunksize=chunk, quoting=3,
                          compression=_compression(smiles_path), on_bad_lines="skip"):
        out.append(df.smiles.to_numpy()[np.isin(df.cid.to_numpy(), keep)])
    smiles = np.concatenate(out) if out else np.array([], dtype=object)
    log(f"pubchem: {len(smiles):,} SMILES read ({time.time() - t0:.0f}s)")
    return smiles


# ---------- PubChem channel: documentation prior and a per-molecule shortlist ----------


def find_pubchem_docs(root="/kaggle/input"):
    """Paths of PubChem's CID-SID and CID-PMID files, or None."""
    found = {}
    for f in glob.glob(os.path.join(root, "**", "*"), recursive=True):
        name = os.path.basename(f).lower()
        if os.path.isfile(f) and name.startswith("cid-sid"):
            found["sid"] = f
        elif os.path.isfile(f) and name.startswith("cid-pmid"):
            found["pmid"] = f
    return found if len(found) == 2 else None


def window_compounds(mass_path, masses, ppm=10.0, chunk=5_000_000):
    """(cid, mono mass) of neutral organic PubChem compounds within ppm of any query mass, sorted by mass."""
    q = np.unique(np.asarray(masses, dtype=float))
    q = q[np.isfinite(q)]
    q = np.r_[q, q[-1]] if len(q) == 1 else q
    cids, monos = [], []
    for df in pd.read_csv(mass_path, sep="\t", header=None, names=["cid", "formula", "mono", "exact"],
                          usecols=["cid", "formula", "mono"],
                          dtype={"cid": np.int64, "formula": str, "mono": np.float64},
                          chunksize=chunk, compression=_compression(mass_path), on_bad_lines="skip"):
        m = df.mono.to_numpy()
        i = np.searchsorted(q, m).clip(1, len(q) - 1)
        sub = df[np.minimum(np.abs(m - q[i]), np.abs(m - q[i - 1])) <= m * ppm * 1e-6]
        sub = sub[sub.formula.str.fullmatch(ORGANIC, na=False)]
        cids.append(sub.cid.to_numpy())
        monos.append(sub.mono.to_numpy())
    cid, mono = np.concatenate(cids), np.concatenate(monos)
    order = np.argsort(mono, kind="stable")
    return cid[order], mono[order]


def count_rows_per_cid(path, cids, chunk=10_000_000):
    """Number of rows per CID in a CID-keyed PubChem file (CID-SID, CID-PMID), aligned with `cids`."""
    uniq = np.unique(cids)
    counts = np.zeros(len(uniq), dtype=np.int64)
    for df in pd.read_csv(path, sep="\t", header=None, usecols=[0], dtype={0: np.int64}, chunksize=chunk,
                          compression=_compression(path), on_bad_lines="skip"):
        c = df[0].to_numpy()
        c = c[np.isin(c, uniq)]
        if len(c):
            v, n = np.unique(c, return_counts=True)
            counts[np.searchsorted(uniq, v)] += n
    return counts[np.searchsorted(uniq, cids)]


def documentation_prior(n_sid, n_pmid):
    """pop = ln(1 + substance records) + ln(1 + PubMed articles): how much a compound has been written about."""
    return np.log1p(n_sid) + np.log1p(n_pmid)


def shortlist_windows(cid, mono, pop, masses, ppm=10.0, depth=1000):
    """Per query mass: the `depth` best-documented compounds of its window (CIDs), and the union of all of them."""
    lists = []
    for x in masses:
        tol = x * ppm * 1e-6
        lo, hi = np.searchsorted(mono, x - tol), np.searchsorted(mono, x + tol, side="right")
        w = np.arange(lo, hi)
        if len(w) > depth:
            w = w[np.argsort(-pop[w], kind="stable")[:depth]]
        lists.append(cid[w])
    return lists, (np.unique(np.concatenate(lists)) if lists else np.array([], dtype=np.int64))


def smiles_for_cids(smiles_path, cids, chunk=5_000_000):
    """{cid: SMILES} for the given CIDs (one pass over CID-SMILES)."""
    want, out = np.unique(cids), {}
    for df in pd.read_csv(smiles_path, sep="\t", header=None, names=["cid", "smiles"],
                          dtype={"cid": np.int64, "smiles": str}, chunksize=chunk, quoting=3,
                          compression=_compression(smiles_path), on_bad_lines="skip"):
        hit = np.isin(df.cid.to_numpy(), want)
        out.update(zip(df.cid.to_numpy()[hit].tolist(), df.smiles.to_numpy()[hit].tolist(), strict=True))
    return out


def convert_cids(cid_smiles, n_bits=2048, workers=None, chunk=20000):
    """RDKit for {cid: SMILES} -> frame[cid, key, smiles, mass, formula] and packed fingerprints (row-aligned)."""
    cids = np.array(list(cid_smiles), dtype=np.int64)
    smiles = [str(cid_smiles[c]) for c in cids]
    jobs = [(smiles[i:i + chunk], n_bits) for i in range(0, len(smiles), chunk)]
    with get_context("fork").Pool(workers or os.cpu_count() or 1) as pool:
        parts = pool.map(_convert, jobs)
    rows, fps = [], []
    for j, p in enumerate(parts):
        base = j * chunk
        rows += [(int(cids[base + k]), key, smi, m, f) for k, key, smi, m, f in zip(p[5], *p[:4], strict=True)]
        fps.append(p[4])
    frame = pd.DataFrame(rows, columns=["cid", "key", "smiles", "mass", "formula"])
    return frame, (np.concatenate(fps) if fps else np.zeros((0, n_bits // 8), dtype=np.uint8))


def merge_channel(main, channel, slots=(), override=False, n=25):
    """Insert channel proposals into a main list at fixed final ranks (1-based `slots`); `override` puts the
    first proposal at rank 1 (the remaining proposals then take the slots). Items are pool/channel ids;
    duplicates are dropped and the list is cut to n."""
    channel = list(channel)
    out, used = [], set()
    take = []
    if override and channel:
        take.append((1, channel.pop(0)))
    take += [(s, c) for s, c in zip(sorted(slots), channel, strict=False)]
    insert_at = dict(take)
    main_iter = iter(main)
    for rank in range(1, n + 1):
        item = insert_at.get(rank)
        if item is None:
            item = next((m for m in main_iter if m not in used), None)
        if item is not None and item not in used:
            out.append(item)
            used.add(item)
    return out


# ---------- precursor mass ----------

WATER = 18.01056468


def consensus_mass(masses, ppm=10.0):
    """Neutral mass most spectra of a molecule agree on (within ppm), not the median.

    With an even number of spectra the median averages the middle two, so one spectrum with an
    unusual or mislabelled adduct can move the mass off every correct candidate.
    """
    m = np.asarray(masses, dtype=float)
    m = m[np.isfinite(m)]
    if len(m) == 0:
        return float("nan")
    tol = float(np.median(m)) * ppm * 1e-6
    support = (np.abs(m[:, None] - m[None, :]) <= tol).sum(axis=1)
    tied = m[support == support.max()]
    centre = tied[np.argmin(np.abs(tied - np.median(m)))]
    return float(m[np.abs(m - centre) <= tol].mean())


# ---------- official-key deduplication of the submission ----------


def _official_keys(smiles_list):
    """Worker: InChIKey14 after RDKit tautomer canonicalisation (how the metric compares answers)."""
    from rdkit import Chem, RDLogger
    from rdkit.Chem.MolStandardize import rdMolStandardize

    RDLogger.DisableLog("rdApp.*")
    enum = rdMolStandardize.TautomerEnumerator()
    enum.SetMaxTautomers(256)  # cap for huge natural products; dedup only, so an approximation is safe
    out = []
    for s in smiles_list:
        try:
            mol = Chem.MolFromSmiles(s)
            out.append(Chem.MolToInchiKey(enum.Canonicalize(mol))[:14] if mol is not None else s)
        except Exception:  # noqa: BLE001
            out.append(s)
    return out


def dedupe_official(rankings, pool, k=25, workers=None, log=print):
    """Drop guesses that the metric would treat as the same structure (tautomers), keep the first k.

    rankings: {molecule: [pool indices]} longer than k, so freed slots are refilled from further down.
    """
    smiles = pool.smiles.to_numpy()
    needed = sorted({int(i) for r in rankings.values() for i in r})
    chunks = [[smiles[i] for i in needed[j:j + 500]] for j in range(0, len(needed), 500)]
    with get_context("fork").Pool(workers or os.cpu_count() or 1) as p:
        keys = [key for part in p.map(_official_keys, chunks) for key in part]
    official = dict(zip(needed, keys, strict=True))
    out, freed = {}, 0
    for mol, ranked in rankings.items():
        seen, kept = set(), []
        for i in ranked[:k]:
            freed += official[int(i)] in seen
            seen.add(official[int(i)])
        seen = set()
        for i in ranked:
            if official[int(i)] not in seen:
                seen.add(official[int(i)])
                kept.append(int(i))
            if len(kept) == k:
                break
        out[mol] = kept
    log(f"official-key dedup: {freed} duplicate slots in the top {k} replaced ({len(needed):,} structures checked)")
    return out


# ---------- gated B2 fusion ----------


def candidate_lib_scores(cp, pool, library, spectra, spectra_mz, spectra_in, cands, modified=True):
    """{molecule: array}: for each candidate, mean over the molecule's spectra of its best library match.

    0 for candidates without reference spectra. cands: {molecule: [pool indices]}.
    """
    keys = pool.key.to_numpy()
    spectra = spectra.assign(pos=np.arange(len(spectra)))
    out = {}
    for mol, group in spectra.groupby("molecule_id", sort=False):
        if mol not in cands:
            continue
        rows = [(r.pos, r.ionization_mode, r.precursor_mz, int((spectra_in[r.pos] > 0).sum()))
                for r in group.itertuples()]
        vals = np.zeros(len(cands[mol]))
        for c, idx in enumerate(cands[mol]):
            per_spec = []
            for pos, mode, prec, n in rows:
                best = 0.0
                for lib_pos in library.index.get((keys[idx], mode), ()) if n else ():
                    m = library.n_peaks[lib_pos]
                    if m == 0:
                        continue
                    shift = (prec - library.precursor[lib_pos]) if modified else None
                    best = max(best, cp.match_score(spectra_mz[pos, :n], spectra_in[pos, :n],
                                                    library.mz[lib_pos, :m], library.intensity[lib_pos, :m],
                                                    shift=shift))
                per_spec.append(best)
            vals[c] = float(np.mean(per_spec)) if per_spec else 0.0
        out[mol] = vals
    return out


def b2_top_scores(cp, pool, library, spectra, spectra_mz, spectra_in, b2_rank, modified=True):
    """Per molecule: the library score of B2's top candidate (the number B2 ranked by, pooling="mean")."""
    top = {mol: [r[0]] for mol, r in b2_rank.items()}
    scores = candidate_lib_scores(cp, pool, library, spectra, spectra_mz, spectra_in, top, modified)
    return {mol: float(v[0]) for mol, v in scores.items()}


def gate(model_rank, alt_rank, scores, tau):
    """alt ranking where the library match is strong (score >= tau), the model's elsewhere."""
    return {m: (alt_rank[m] if scores.get(m, 0.0) >= tau else r) for m, r in model_rank.items()}


def _balanced(cp, rankings, truth, pool_keys):
    return float(cp.evaluate(rankings, truth, pool_keys).loc["balanced", "MRR@25"])


def tune_gate(cp, truth, pool_keys, model_rank, alts, scores, taus=None):
    """Pick (alt, tau) on validation by balanced MRR@25.

    alts: {name: rankings} to switch to (e.g. B2, fusion). Returns a dict with the chosen alt and tau,
    the in-sample score, and a 2-fold cross-fitted score (tau tuned on one half, applied to the other).
    """
    s = np.array([scores.get(m, 0.0) for m in truth.molecule_id])
    taus = taus if taus is not None else np.unique(np.r_[np.quantile(s[s > 0], np.linspace(0, 1, 41)) if
                                                          (s > 0).any() else [], np.inf])
    fold = np.array([zlib.crc32(m.encode()) % 2 for m in truth.molecule_id])

    def best_on(mask):
        sub = truth[mask]
        best = (-1.0, None, np.inf)
        for name, alt in alts.items():
            for tau in taus:
                v = _balanced(cp, gate(model_rank, alt, scores, tau), sub, pool_keys)
                if v > best[0]:
                    best = (v, name, tau)
        return best

    score_in, name, tau = best_on(np.ones(len(truth), dtype=bool))
    crossed = {}
    for f in (0, 1):
        _, n_f, t_f = best_on(fold == f)
        held = truth.molecule_id[fold != f]
        g = gate({m: model_rank[m] for m in held}, alts[n_f], scores, t_f)
        crossed.update(g)
    return {"alt": name, "tau": float(tau), "in_sample": round(score_in, 4),
            "cross_fitted": round(_balanced(cp, crossed, truth, pool_keys), 4),
            "share_gated_val": round(float((s >= tau).mean()), 4)}
```

```python
%%writefile m7/common/formula.py
"""Fragment-formula explanation: can a candidate's molecular formula produce the observed fragment peaks?

Every fragment is made of atoms the molecule has. For a candidate formula we enumerate its heavy-atom
sub-formulas (C, N, O, P, S, halogens), allow any hydrogen count up to the parent's, and check whether
each strong peak matches one of them as a protonated / deprotonated / radical ion (or keeps the Na / K
of a metal adduct). The score is the share of peak intensity explained, in [0, 1].

Candidates that share a formula share the score, so a window of ~8,000 PubChem compounds needs only
the few dozen distinct formulas in it. Wrong-formula candidates (same mass, different atoms) score low;
true isomers of the answer score the same as the answer and are left to the fingerprint model.
"""

from __future__ import annotations

import re
from functools import lru_cache

import numpy as np

H = 1.00782503207
PROTON = 1.00727646688
ELECTRON = 0.00054857990907
NA, K = 22.98976928, 38.96370649
HEAVY = {"C": 12.0, "N": 14.0030740048, "O": 15.99491461956, "P": 30.97376163, "S": 31.972071,
         "F": 18.99840322, "Cl": 34.96885268, "Br": 78.9183371, "I": 126.904473}
MAX_SUBFORMULAS = 400_000  # guard for very large formulas


def parse_formula(formula):
    """'C10H15ClN2' -> {'C': 10, 'H': 15, 'Cl': 1, 'N': 2}; None for charged or unsupported formulas."""
    f = str(formula)
    if not f or any(c in f for c in "+-.") or not re.fullmatch(r"(?:[A-Z][a-z]?\d*)+", f):
        return None
    counts = {}
    for el, n in re.findall(r"([A-Z][a-z]?)(\d*)", f):
        if el != "H" and el not in HEAVY:
            return None
        counts[el] = counts.get(el, 0) + (int(n) if n else 1)
    return counts


@lru_cache(maxsize=200_000)
def _heavy_masses(formula):
    """Sorted masses of every heavy-atom sub-formula, and the parent's hydrogen count."""
    counts = parse_formula(formula)
    if counts is None:
        return None, 0
    masses = np.zeros(1)
    for el, m in HEAVY.items():
        n = counts.get(el, 0)
        if n:
            masses = (masses[:, None] + np.arange(n + 1) * m).ravel()
            if len(masses) > MAX_SUBFORMULAS:
                return None, 0
    return np.sort(masses), counts.get("H", 0)


def ion_shifts(adduct, mode):
    """Neutral-fragment mass = m/z - shift. Protonated or radical ions, plus the adduct's metal if any."""
    a = str(adduct)
    if str(mode).startswith("neg") or a.endswith("-"):
        return [-PROTON, ELECTRON]
    shifts = [PROTON, -ELECTRON]
    if "Na" in a:
        shifts.append(NA - ELECTRON)
    if "K" in a and "NH4" not in a:
        shifts.append(K - ELECTRON)
    return shifts


def molecule_peaks(rows, mz, intensity, adducts, modes, top=20, floor=0.02):
    """Strong peaks of one molecule's spectra: (neutral-fragment masses per shift, weights).

    rows: positions of the molecule's spectra in mz / intensity. Each spectrum's top peaks get equal total
    weight, so a molecule with many spectra is not dominated by one of them.
    """
    targets, weights = [], []
    for r in rows:
        inten = intensity[r]
        n = int((inten > 0).sum())
        if n == 0:
            continue
        x, w = mz[r, :n], inten[:n]
        keep = np.argsort(-w)[:top]
        keep = keep[w[keep] >= floor * w.max()]
        x, w = x[keep], w[keep] / w[keep].sum()
        shifts = np.array(ion_shifts(adducts[r], modes[r]))
        targets.append(x[:, None] - shifts[None, :])
        weights.append(w)
    if not targets:
        return None
    width = max(t.shape[1] for t in targets)  # pad shift columns with NaN so spectra stack
    targets = [np.pad(t, ((0, 0), (0, width - t.shape[1])), constant_values=np.nan) for t in targets]
    w = np.concatenate(weights)
    return np.concatenate(targets), w / w.sum()


def explained_fraction(formula, peaks, ppm=10.0, abs_tol=0.002):
    """Share of peak weight explained by sub-formulas of `formula` (nan if the formula is unusable)."""
    hm, hmax = _heavy_masses(formula)
    if hm is None or peaks is None:
        return np.nan
    targets, weights = peaks
    ok = np.zeros(len(targets), dtype=bool)
    for j in range(targets.shape[1]):
        t = targets[:, j]
        valid = np.isfinite(t)
        if not valid.any():
            continue
        tol = np.maximum(abs_tol, t[valid] * ppm * 1e-6)
        # for each target, every heavy sub-formula lighter than it; the rest must be hydrogens
        hi = np.searchsorted(hm, t[valid] + tol, side="right")
        for k, (tv, tl, h_end) in enumerate(zip(t[valid], tol, hi, strict=True)):
            res = tv - hm[:h_end]
            nh = np.rint(res / H)
            hit = (nh >= 0) & (nh <= hmax + 1) & (np.abs(res - nh * H) <= tl)
            if hit.any():
                ok[np.flatnonzero(valid)[k]] = True
    return float(weights[ok].sum())


def window_scores(formulas, peaks, cache=None, **kw):
    """Explanation score for each candidate in a window (candidates with the same formula share it)."""
    cache = {} if cache is None else cache
    out = np.empty(len(formulas))
    for i, f in enumerate(formulas):
        if f not in cache:
            cache[f] = explained_fraction(f, peaks, **kw)
        out[i] = cache[f]
    return out


def has_elements(formula, elements=("F", "I")):
    counts = parse_formula(formula)
    return bool(counts) and any(counts.get(e, 0) for e in elements)
```

```python
%%writefile m7/common/rerank.py
"""Second-stage reranker: learn to pick the answer out of its own line-up of same-mass candidates.

The first stage (fingerprint model + soft Tanimoto) puts the answer in the top 25 for ~82% of validation
molecules but at rank 1 for only ~49%. This module re-orders each molecule's top-K with a learning-to-rank
model trained on line-ups: one group per validation molecule, label 1 for the answer, 0 for its same-mass
competitors. Features are spectrum-versus-structure evidence only:

  fp   fingerprint likelihood per method (not just similarity), confident-bit conflicts, list-relative scores
  ms   exact-mass evidence: fragment-formula explanation, candidate mass error, neutral losses seen in the
       spectrum (water, CO2, sugars, methoxy, ...) matched against the candidate's functional groups
  bond fragments from breaking 1-2 bonds of the candidate's actual structure, matched to the peaks
       (isomers share a formula but break differently)
  lib  how well the candidate's reference spectra in the library match (0 when it has none)
  base the first-stage score and rank

No provenance features (training library, PubChem-only, enveda-180): validation answers are always
training-library structures, so such features would teach "the answer is never PubChem-only".
The one database signal is the PubChem documentation prior (DOC_FEATURES, v8): it is its own ablation
step, and the notebook uses it only if it also lifts the natural-product examples, because library
standards are better documented than the average hidden answer (v8 scored 0.276 vs v7's 0.281 with it, so v9
leaves it off). CLIP_FEATURES (v9) are the contrastive transformer's scores: spectrum-versus-structure evidence
like the rest, added as their own ablation step.
"""

from __future__ import annotations

import os
import zlib
from functools import lru_cache
from multiprocessing import get_context

import numpy as np
import pandas as pd

from common.formula import H, explained_fraction, ion_shifts

# neutral losses: (mass, SMARTS a candidate needs to show this loss)
LOSSES = {
    "h2o": (18.010565, "[OX2H1]"),
    "co2": (43.989829, "[CX3](=O)[OX2H1,OX1-]"),
    "hcooh": (46.005479, "[CX3](=O)[OX2H1,OX1-]"),
    "nh3": (17.026549, "[NX3;H2,H1;!$(NC=O)]"),
    "ch3oh": (32.026215, "[CH3][OX2]"),
    "c2h2o": (42.010565, "[CH3]C(=O)[O,N]"),
    # pyranose rings: anomeric C1(-O), ring O, C5 carrying CH2OH / CH3 / ring CH2 / COOH, C2-C4 each with an O
    "hexose": (162.052824, "[CX4;R]1([OX2])[OX2;R][CX4;R]([CH2][OX2])[CX4;R]([OX2])[CX4;R]([OX2])[CX4;R]1[OX2]"),
    "dhex": (146.057909, "[CX4;R]1([OX2])[OX2;R][CX4;R]([CH3])[CX4;R]([OX2])[CX4;R]([OX2])[CX4;R]1[OX2]"),
    "pentose": (132.042259, "[CX4;R]1([OX2])[OX2;R][CH2;R][CX4;R]([OX2])[CX4;R]([OX2])[CX4;R]1[OX2]"),
    "glca": (176.032088, "[CX4;R]1([OX2])[OX2;R][CX4;R](C(=O)[OX2H1,OX1-])[CX4;R]([OX2])[CX4;R]([OX2])[CX4;R]1[OX2]"),
    "so3": (79.956815, "S(=O)(=O)[OX2H1,OX1-,OX2]"),
    "co": (27.994915, "[CX3]=O"),
}
GROUPS = {
    "base": ["first_score", "first_rank"],
    "fp": [],  # filled below (per method)
    "ms": ["expl_2mda", "expl_5mda", "mass_err_ppm", "loss_agree", "loss_conflict"]
          + [f"loss_{k}" for k in LOSSES],
    "bond": ["bond_expl", "bond_expl_1cut", "bond_top5"],
    "lib": ["lib_score"],
}
METHODS = ("mean_tuned", "mean", "attention_head")
for _m in METHODS:
    GROUPS["fp"] += [f"{_m}_tan", f"{_m}_ll", f"{_m}_miss_on", f"{_m}_extra_on", f"{_m}_tan_gap", f"{_m}_ll_gap"]
ALL_FEATURES = [f for g in GROUPS.values() for f in g]


# ---------- spectrum side ----------


def spectrum_evidence(rows, mz, intensity, precursors, adducts, modes, top=30, tol=0.005):
    """Peaks (as neutral-fragment targets) and the neutral losses a molecule's spectra show."""
    targets, weights, losses = [], [], np.zeros(len(LOSSES))
    loss_masses = np.array([v[0] for v in LOSSES.values()])
    for r in rows:
        inten = intensity[r]
        n = int((inten > 0).sum())
        if n == 0:
            continue
        x, w = mz[r, :n], inten[:n]
        keep = np.argsort(-w)[:top]
        x, w = x[keep], w[keep] / w[keep].sum()
        shifts = np.array(ion_shifts(adducts[r], modes[r]))
        targets.append(np.pad(x[:, None] - shifts[None, :], ((0, 0), (0, 4 - len(shifts))), constant_values=np.nan))
        weights.append(w)
        diffs = np.abs((precursors[r] - x)[:, None] - loss_masses[None, :]) <= tol  # precursor -> fragment losses
        pair = np.abs(np.abs(x[:, None] - x[None, :])[..., None] - loss_masses) <= tol  # fragment -> fragment
        losses = np.maximum(losses, (diffs.any(0) | pair.any((0, 1))).astype(float))
    if not targets:
        return None
    w = np.concatenate(weights)
    return {"targets": np.concatenate(targets), "weights": w / w.sum(), "losses": losses}


# ---------- candidate side ----------


@lru_cache(maxsize=1)
def _loss_patterns():
    from rdkit import Chem

    return [Chem.MolFromSmarts(v[1]) for v in LOSSES.values()]


@lru_cache(maxsize=400_000)
def candidate_structure(smiles, max_pairs=200):
    """(sorted fragment masses with cut counts, functional-group vector) for one SMILES, or None.

    Fragments: the pieces left after removing one acyclic bond, two bonds of the same ring, or two acyclic
    bonds (capped). Each atom keeps its implicit hydrogens; matching allows -2..+2 H shifts.
    """
    from rdkit import Chem

    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return None
    n = mol.GetNumAtoms()
    table = Chem.GetPeriodicTable()
    atom_mass = np.array([table.GetMostCommonIsotopeMass(a.GetSymbol()) + a.GetTotalNumHs() * H
                          for a in mol.GetAtoms()])
    adj = [[] for _ in range(n)]
    bonds = []
    for b in mol.GetBonds():
        i, j = b.GetBeginAtomIdx(), b.GetEndAtomIdx()
        adj[i].append((j, b.GetIdx()))
        adj[j].append((i, b.GetIdx()))
        bonds.append((b.GetIdx(), b.IsInRing()))

    def pieces(cut):
        seen, out = np.zeros(n, dtype=bool), []
        for s in range(n):
            if seen[s]:
                continue
            stack, total = [s], 0.0
            seen[s] = True
            while stack:
                a = stack.pop()
                total += atom_mass[a]
                for nb, bi in adj[a]:
                    if bi not in cut and not seen[nb]:
                        seen[nb] = True
                        stack.append(nb)
            out.append(total)
        return out if len(out) > 1 else []

    masses, cuts = [float(atom_mass.sum())], [0]
    acyclic = [bi for bi, ring in bonds if not ring]
    for bi in acyclic:
        for m in pieces({bi}):
            masses.append(m)
            cuts.append(1)
    for ring in mol.GetRingInfo().BondRings():
        ring = list(ring)
        for a in range(len(ring)):
            for b in range(a + 1, len(ring)):
                for m in pieces({ring[a], ring[b]}):
                    masses.append(m)
                    cuts.append(2)
    pairs = [(a, b) for k, a in enumerate(acyclic) for b in acyclic[k + 1:]][:max_pairs]
    for a, b in pairs:
        for m in pieces({a, b}):
            masses.append(m)
            cuts.append(2)
    # keep each distinct piece mass once (fewest cuts): pieces repeat a lot, and this bounds memory
    masses, cuts = np.round(np.asarray(masses), 4), np.asarray(cuts, dtype=np.int8)
    order = np.lexsort((cuts, masses))
    masses, cuts = masses[order], cuts[order]
    first = np.r_[True, masses[1:] != masses[:-1]]
    groups = np.array([float(mol.HasSubstructMatch(p)) if p is not None else 0.0 for p in _loss_patterns()])
    return masses[first], cuts[first], groups


def bond_explanation(structure, evidence, tol=0.005):
    """Share of peak weight explained by bond-cleavage pieces (any cuts / one cut), and of the top-5 peaks."""
    if structure is None or evidence is None:
        return np.nan, np.nan, np.nan
    masses, cuts, _ = structure
    t, w = evidence["targets"], evidence["weights"]
    hit_any = np.zeros(len(t), dtype=bool)
    hit_one = np.zeros(len(t), dtype=bool)
    one = masses[cuts <= 1]
    for dh in range(-2, 3):
        for col in range(t.shape[1]):
            x = t[:, col] - dh * H
            ok = np.isfinite(x)
            for arr, hit in ((masses, hit_any), (one, hit_one)):
                if len(arr) == 0:
                    continue
                i = np.searchsorted(arr, x[ok]).clip(1, len(arr) - 1)
                d = np.minimum(np.abs(arr[i] - x[ok]), np.abs(arr[i - 1] - x[ok]))
                hit[np.flatnonzero(ok)[d <= tol]] = True
    top5 = np.argsort(-w)[:5]
    return float(w[hit_any].sum()), float(w[hit_one].sum()), float(hit_any[top5].mean())


def _structures(smiles_list):
    return [candidate_structure(s) for s in smiles_list]


def precompute_structures(smiles, workers=None):
    """Fragment tables for many SMILES in parallel (fills each worker's cache, returns a dict)."""
    uniq = list(dict.fromkeys(smiles))
    chunks = [uniq[i:i + 500] for i in range(0, len(uniq), 500)]
    with get_context("fork").Pool(workers or os.cpu_count() or 1) as p:
        parts = p.map(_structures, chunks)
    return dict(zip(uniq, [s for part in parts for s in part], strict=True))


# ---------- features for one molecule's line-up ----------


def fp_features(probs, bits):
    """probs: (F,) pooled probabilities; bits: (K, F) candidate fingerprints -> dict of (K,) arrays."""
    p = np.clip(probs.astype(np.float64), 1e-4, 1 - 1e-4)
    b = bits.astype(np.float64)
    overlap = b @ p
    tan = overlap / (p.sum() + b.sum(1) - overlap + 1e-8)
    ll = (b @ np.log(p) + (1 - b) @ np.log(1 - p)) / len(p)
    return {"tan": tan, "ll": ll, "miss_on": ((p > 0.8) & (b == 0)).sum(1).astype(float),
            "extra_on": ((p < 0.05) & (b == 1)).sum(1).astype(float), "tan_gap": tan - tan.max(),
            "ll_gap": ll - ll.max()}


def lineup_features(cand, first_score, probs_by_method, pool_fp, pool_smiles, pool_formula, pool_mass, mass,
                    evidence, structures, lib_scores):
    """Feature table (K rows) for one molecule's candidates `cand` (pool indices, first-stage order)."""
    k = len(cand)
    bits = np.unpackbits(pool_fp[cand], axis=1)
    f = {"first_score": np.asarray(first_score, dtype=float), "first_rank": np.arange(k, dtype=float)}
    for m in METHODS:
        for name, v in fp_features(probs_by_method[m], bits).items():
            f[f"{m}_{name}"] = v
    peaks = None if evidence is None else (evidence["targets"], evidence["weights"])
    cache = {}
    for name, tol in (("expl_2mda", 0.002), ("expl_5mda", 0.005)):
        vals = []
        for fo in pool_formula[cand]:
            key = (fo, tol)
            if key not in cache:
                cache[key] = explained_fraction(fo, peaks, abs_tol=tol)
            vals.append(cache[key])
        f[name] = np.array(vals, dtype=float)
    f["mass_err_ppm"] = np.abs(pool_mass[cand] - mass) / mass * 1e6
    spec_loss = evidence["losses"] if evidence is not None else np.zeros(len(LOSSES))
    groups = np.array([structures[s][2] if structures.get(s) is not None else np.full(len(LOSSES), np.nan)
                       for s in pool_smiles[cand]])
    f["loss_agree"] = np.nansum(groups * spec_loss, 1)
    f["loss_conflict"] = np.nansum((1 - groups) * spec_loss, 1)
    for j, name in enumerate(LOSSES):
        f[f"loss_{name}"] = groups[:, j] * spec_loss[j]
    be = np.array([bond_explanation(structures.get(s), evidence) for s in pool_smiles[cand]])
    f["bond_expl"], f["bond_expl_1cut"], f["bond_top5"] = be[:, 0], be[:, 1], be[:, 2]
    f["lib_score"] = np.asarray(lib_scores, dtype=float)
    return pd.DataFrame(f)[ALL_FEATURES]


# ---------- learning to rank ----------


LGB_PARAMS = {"objective": "lambdarank", "learning_rate": 0.05, "num_leaves": 31, "min_data_in_leaf": 20,
              "bagging_fraction": 0.8, "bagging_freq": 1, "feature_fraction": 0.8, "lambdarank_truncation_level": 25,
              "verbose": -1}


def fit(table, features, seed=0, rounds=400):
    """LambdaRank on line-ups. table: rows grouped by 'group' (contiguous), columns features + 'label'."""
    import lightgbm as lgb

    sizes = table.groupby("group", sort=False).size().to_numpy()
    data = lgb.Dataset(table[features].to_numpy(np.float32), table["label"].to_numpy(), group=sizes,
                       feature_name=list(features), free_raw_data=False)
    return lgb.train({**LGB_PARAMS, "seed": seed}, data, num_boost_round=rounds)


def fit_bag(table, features, seeds=(0, 1, 2), rounds=400):
    """Several seeds of `fit` (row and feature bagging differ), averaged by `predict`."""
    return [fit(table, features, seed, rounds) for seed in seeds]


def predict(model, table, features):
    x = table[features].to_numpy(np.float32)
    if isinstance(model, list):
        return np.mean([m.predict(x) for m in model], axis=0)
    return model.predict(x)


# ---------- documentation prior (kept out of ALL_FEATURES: it is its own ablation step) ----------
DOC_FEATURES = ["doc_pop", "doc_known", "doc_rank", "doc_gap", "doc_iso_gap"]


def doc_features(keys, formulas, key_pop):
    """How well documented each line-up candidate is in PubChem, absolute and relative to its competitors.

    key_pop: {inchikey14: pop} for the best-documented PubChem compounds of this molecule's mass window.
    A candidate missing from it is less documented than all of them (or not in PubChem): doc_pop is nan,
    doc_known 0. doc_iso_gap compares a candidate with the best-documented candidate of the same formula,
    which is the comparison that matters when the answer loses to an isomer.
    """
    pop = np.array([key_pop.get(k, np.nan) for k in keys], dtype=float)
    known = np.isfinite(pop)
    filled = np.where(known, pop, -1.0)
    rank = np.empty(len(pop))
    rank[np.argsort(-filled, kind="stable")] = np.arange(1, len(pop) + 1)
    rank[~known] = len(pop) + 1
    top = filled.max() if len(pop) else -1.0
    iso_top = pd.Series(filled).groupby(np.asarray(formulas, dtype=object)).transform("max").to_numpy()
    return pd.DataFrame({
        "doc_pop": pop,
        "doc_known": known.astype(np.float32),
        "doc_rank": rank,
        "doc_gap": np.where(known, pop - top, np.nan),
        "doc_iso_gap": np.where(known, pop - iso_top, np.nan),
    })


# ---------- contrastive transformer (Architecture 3, common/contrastive.py; its own ablation step) ----------
CLIP_FEATURES = ["clip_sim", "clip_rank", "clip_gap", "clip_iso_gap", "clip_z"]


def clip_features(sim, formulas):
    """Contrastive score of each line-up candidate, absolute and relative to the line-up and to its isomers."""
    sim = np.asarray(sim, dtype=float)
    rank = np.empty(len(sim))
    rank[np.argsort(-sim, kind="stable")] = np.arange(1, len(sim) + 1)
    iso_top = pd.Series(sim).groupby(np.asarray(formulas, dtype=object)).transform("max").to_numpy()
    top = sim.max() if len(sim) else 0.0
    return pd.DataFrame({
        "clip_sim": sim,
        "clip_rank": rank,
        "clip_gap": sim - top,
        "clip_iso_gap": sim - iso_top,
        "clip_z": (sim - sim.mean()) / (sim.std() + 1e-6) if len(sim) else sim,
    })


def folds(groups, n=5):
    return np.array([zlib.crc32(str(g).encode()) % n for g in groups])


def cross_fit(table, features, n_folds=5, seed=0):
    """Out-of-fold scores: each molecule is scored by a model that never saw its line-up."""
    out = np.full(len(table), np.nan)
    fold = folds(table["group"].to_numpy(), n_folds)
    for f in range(n_folds):
        train = table[fold != f]
        if train.label.sum() == 0 or (fold == f).sum() == 0:
            continue
        fitted = fit(train, features, seed)
        out[fold == f] = predict(fitted, table[fold == f], features)
    return out
```

```python
%%writefile m7/common/contrastive.py
"""Architecture 3: contrastive spectrum-molecule retrieval with a peak transformer. Owner: Member 7.

Why: after reranking, 70-80% of the validation MRR we still lose is the answer beaten by a same-formula
isomer. The first stage predicts a 2,048-bit fingerprint and compares it with Tanimoto, and close isomers share
almost every bit, so that comparison cannot separate them however good the encoder is. This model is trained to
do exactly that job: pick the right structure out of the candidates that share its mass.

    spectrum peaks --PeakEncoder (transformer)--> h_i  (one per spectrum)
    {h_1..h_n}     --GatedAttentionPool--------> q    (one per molecule; starts as the mean, learns weights)
    candidate      --MoleculeEncoder (MLP on Morgan bits)--> m
    score(q, m) = cos(q, m) * scale                        (CLIP-style, learned temperature)

Training (InfoNCE): for each training molecule the positive is its own structure; the negatives are the other
molecules in the batch plus ``n_neg`` structures drawn from its own +-ppm mass window of the candidate pool, which
are mostly same-formula isomers (hard negatives). A small fingerprint head on q (BCE) keeps the embedding
chemically grounded early in training.

Peaks are tokens (exact m/z and neutral loss as Fourier features, plus intensity), so nothing is binned. The pool
uses Ilse et al.'s gated attention with the scoring layer initialised to zero: it is exactly mean pooling at the
start and only moves away from it if that helps the contrastive loss, which is the fair test of whether learned
pooling beats the mean when the encoder is trained with it (Member 7's Arch 1 / 2 only had a frozen encoder).
"""

from __future__ import annotations

import contextlib
import math
import time

import numpy as np
import pandas as pd
import torch
import torch.nn.functional as F
from torch import nn

from common.sets import pad_sets, sample_sets

N_ADDUCTS_DEFAULT = 16


# ---------- data ----------


def top_peaks(mz: np.ndarray, intensity: np.ndarray, n: int = 64,
              chunk: int = 200_000) -> tuple[np.ndarray, np.ndarray]:
    """Keep the ``n`` most intense peaks of each padded row (padding = intensity 0), sorted by m/z."""
    if mz.shape[1] <= n:
        return mz.astype(np.float32), intensity.astype(np.float32)
    out_mz = np.zeros((len(mz), n), np.float32)
    out_in = np.zeros((len(mz), n), np.float32)
    for s in range(0, len(mz), chunk):  # chunks: 1.7M training spectra would need ~4 GB of temporaries at once
        m, it = mz[s:s + chunk], intensity[s:s + chunk]
        keep = np.argpartition(-it, n - 1, axis=1)[:, :n]
        mz_k, in_k = np.take_along_axis(m, keep, 1), np.take_along_axis(it, keep, 1)
        order = np.argsort(np.where(in_k > 0, mz_k, np.inf), axis=1, kind="stable")
        mz_k, in_k = np.take_along_axis(mz_k, order, 1), np.take_along_axis(in_k, order, 1)
        mz_k[in_k <= 0] = 0.0
        out_mz[s:s + chunk], out_in[s:s + chunk] = mz_k, in_k
    return out_mz, out_in


def window_bounds(pool_mass: np.ndarray, masses: np.ndarray, ppm: float) -> tuple[np.ndarray, np.ndarray]:
    """[lo, hi) of the +-ppm window of each mass in a mass-sorted pool."""
    masses = np.asarray(masses, dtype=np.float64)
    tol = masses * ppm * 1e-6
    return (np.searchsorted(pool_mass, masses - tol, side="left"),
            np.searchsorted(pool_mass, masses + tol, side="right"))


def sample_negatives(rng: np.random.Generator, pos: np.ndarray, lo: np.ndarray, hi: np.ndarray, n_neg: int,
                     n_pool: int) -> np.ndarray:
    """(B, n_neg) pool indices from each positive's own mass window, never the positive itself.

    A window with no other structure falls back to random pool structures (easy negatives).
    """
    pos, lo, hi = (np.asarray(a, dtype=np.int64) for a in (pos, lo, hi))
    width = hi - lo  # includes the positive
    others = np.maximum(width - 1, 0)
    u = rng.random((len(pos), n_neg))
    r = lo[:, None] + np.floor(u * np.maximum(others, 1)[:, None]).astype(np.int64)
    r = np.where(r >= pos[:, None], r + 1, r)  # skip the positive
    rand = rng.integers(0, n_pool, size=r.shape)
    rand = np.where(rand == pos[:, None], (rand + 1) % n_pool, rand)
    return np.where((others > 0)[:, None], r, rand)


# ---------- model ----------


class PeakEncoder(nn.Module):
    """One spectrum -> one vector. Tokens: a summary token (precursor + adduct) and one token per peak."""

    def __init__(self, d: int = 256, layers: int = 4, heads: int = 8, dropout: float = 0.1, n_freq: int = 32,
                 n_adducts: int = N_ADDUCTS_DEFAULT):
        super().__init__()
        # wavelengths from 0.01 Da (separates fragments of nearly equal mass) to 1,000 Da (coarse position)
        self.register_buffer("freqs", 2 * math.pi / torch.logspace(-2.0, 3.0, n_freq))
        self.peak = nn.Sequential(nn.Linear(4 * n_freq + 2, d), nn.GELU(), nn.Linear(d, d))
        self.summary = nn.Sequential(nn.Linear(2 * n_freq, d), nn.GELU(), nn.Linear(d, d))
        self.adduct = nn.Embedding(n_adducts, d)
        layer = nn.TransformerEncoderLayer(d, heads, 4 * d, dropout, activation="gelu", batch_first=True,
                                           norm_first=True)
        self.encoder = nn.TransformerEncoder(layer, layers, enable_nested_tensor=False)
        self.norm = nn.LayerNorm(d)

    def fourier(self, value: torch.Tensor) -> torch.Tensor:
        angle = value.unsqueeze(-1) * self.freqs
        return torch.cat([angle.sin(), angle.cos()], dim=-1)

    def forward(self, mz, intensity, precursor, adduct):
        loss = precursor.unsqueeze(1) - mz  # neutral loss of each peak
        rel = intensity / intensity.amax(1, keepdim=True).clamp(min=1e-6)
        tok = self.peak(torch.cat([self.fourier(mz), self.fourier(loss), intensity.unsqueeze(-1),
                                   rel.unsqueeze(-1)], dim=-1))
        summ = (self.summary(self.fourier(precursor)) + self.adduct(adduct.clamp(0, self.adduct.num_embeddings - 1)))
        pad = torch.cat([torch.zeros_like(intensity[:, :1], dtype=torch.bool), intensity <= 0], dim=1)
        out = self.encoder(torch.cat([summ.unsqueeze(1), tok], dim=1), src_key_padding_mask=pad)
        return self.norm(out[:, 0])


class GatedAttentionPool(nn.Module):
    """Gated attention over a set (Ilse et al., 2018); identical to the mean at initialisation."""

    def __init__(self, d: int, att: int = 128):
        super().__init__()
        self.V, self.U, self.w = nn.Linear(d, att), nn.Linear(d, att), nn.Linear(att, 1)
        nn.init.zeros_(self.w.weight)
        nn.init.zeros_(self.w.bias)

    def forward(self, h: torch.Tensor, mask: torch.Tensor, mode: str = "attention"):
        if mode == "mean":
            m = mask.unsqueeze(-1).to(h.dtype)
            return (h * m).sum(1) / m.sum(1).clamp(min=1), None
        s = self.w(torch.tanh(self.V(h)) * torch.sigmoid(self.U(h))).squeeze(-1).float()
        s = s.masked_fill(~mask, float("-inf"))
        a = torch.softmax(s, dim=1)
        return (a.unsqueeze(-1).to(h.dtype) * h).sum(1), a


class MoleculeEncoder(nn.Module):
    """Candidate structure (Morgan bits) -> vector."""

    def __init__(self, fp_bits: int = 2048, hidden: int = 1024, d: int = 256, dropout: float = 0.1):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(fp_bits, hidden), nn.GELU(), nn.Dropout(dropout), nn.LayerNorm(hidden),
                                 nn.Linear(hidden, hidden), nn.GELU(), nn.Dropout(dropout), nn.Linear(hidden, d))

    def forward(self, bits: torch.Tensor) -> torch.Tensor:
        return self.net(bits)


class Retriever(nn.Module):
    def __init__(self, d: int = 256, layers: int = 4, heads: int = 8, emb: int = 256, fp_bits: int = 2048,
                 n_adducts: int = N_ADDUCTS_DEFAULT, dropout: float = 0.1):
        super().__init__()
        self.spectra = PeakEncoder(d, layers, heads, dropout, n_adducts=n_adducts)
        self.pool = GatedAttentionPool(d)
        self.mol = MoleculeEncoder(fp_bits, 4 * emb, d, dropout)
        self.q_proj, self.m_proj = nn.Linear(d, emb), nn.Linear(d, emb)
        self.fp_head = nn.Linear(d, fp_bits)
        self.log_scale = nn.Parameter(torch.tensor(math.log(1 / 0.07)))

    def scale(self) -> torch.Tensor:
        return self.log_scale.clamp(max=math.log(100.0)).exp()

    def query(self, h: torch.Tensor, mask: torch.Tensor, mode: str = "attention"):
        pooled, att = self.pool(h, mask, mode)
        return F.normalize(self.q_proj(pooled).float(), dim=-1), self.fp_head(pooled), att

    def molecules(self, bits: torch.Tensor) -> torch.Tensor:
        return F.normalize(self.m_proj(self.mol(bits)).float(), dim=-1)


def info_nce(q: torch.Tensor, m_pos: torch.Tensor, m_neg: torch.Tensor, scale: torch.Tensor) -> torch.Tensor:
    """q, m_pos: (B, E); m_neg: (B, H, E). Each anchor must pick its own positive over the batch's other
    positives and its own hard negatives."""
    logits = torch.cat([q @ m_pos.T, torch.einsum("be,bhe->bh", q, m_neg)], dim=1) * scale
    return F.cross_entropy(logits, torch.arange(len(q), device=q.device))


def _unpack(packed: torch.Tensor) -> torch.Tensor:
    shifts = torch.tensor([7, 6, 5, 4, 3, 2, 1, 0], device=packed.device, dtype=torch.uint8)
    return ((packed.unsqueeze(-1) >> shifts) & 1).flatten(-2).float()


# ---------- training ----------


def train_retriever(model, spec, groups, train_keys, key_index, pool_fp, pool_mass, device, *, epochs=15,
                    batch=256, n_neg=31, lr=3e-4, max_set=8, ppm=10.0, aux_weight=0.3, max_minutes=100.0,
                    n_gpu=1, seed=0, score_fn=None, log=print):
    """Train ``model`` with InfoNCE on training molecules; keep the epoch with the best ``score_fn(model)``.

    spec: dict of per-row training spectra arrays mz, intensity (N, P), precursor (N,), adduct (N,).
    groups: {key: row indices into spec}; pool_mass must be sorted (Member 3's pool is).
    """
    torch.manual_seed(seed)
    rng = np.random.default_rng(seed)
    model = model.to(device)
    spec_net = nn.DataParallel(model.spectra, device_ids=list(range(n_gpu))) if n_gpu > 1 else model.spectra
    fp_dev = torch.as_tensor(pool_fp, device=device)
    t = {k: torch.as_tensor(v) for k, v in spec.items()}
    t["precursor"] = t["precursor"].float()
    keys = list(train_keys)
    pos_all = np.array([key_index[k] for k in keys], dtype=np.int64)
    lo_all, hi_all = window_bounds(pool_mass, pool_mass[pos_all], ppm)
    steps_ep = max(len(keys) // batch, 1)
    opt = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=1e-4)
    sched = torch.optim.lr_scheduler.OneCycleLR(opt, max_lr=lr, total_steps=max(epochs * steps_ep, 2),
                                                pct_start=0.05)
    use_amp = str(device).startswith("cuda")
    scaler = torch.amp.GradScaler("cuda", enabled=use_amp)
    autocast = (lambda: torch.autocast("cuda", dtype=torch.float16)) if use_amp else contextlib.nullcontext
    start, best, best_state, hist = time.time(), -1.0, None, []
    for ep in range(epochs):
        model.train()
        order = rng.permutation(len(keys))
        tot = {"nce": 0.0, "bce": 0.0, "n": 0}
        for s in range(0, len(order) - batch + 1, batch):
            sel = order[s:s + batch]
            sets = sample_sets(groups, [keys[j] for j in sel], rng, max_set=max_set)
            rows = np.concatenate(sets)
            local = np.split(np.arange(len(rows)), np.cumsum([len(x) for x in sets])[:-1])
            neg = sample_negatives(rng, pos_all[sel], lo_all[sel], hi_all[sel], n_neg, len(pool_mass))
            r = torch.as_tensor(rows)
            with autocast():
                h = spec_net(t["mz"][r].to(device), t["intensity"][r].to(device), t["precursor"][r].to(device),
                             t["adduct"][r].to(device))
                hs, mask = pad_sets(local, h)
                q, fp_logits, _ = model.query(hs, mask)
                idx = torch.as_tensor(np.concatenate([pos_all[sel][:, None], neg], 1), device=device)
                bits = _unpack(fp_dev[idx.flatten()])
                m = model.molecules(bits).view(len(sel), 1 + n_neg, -1)
                nce = info_nce(q, m[:, 0], m[:, 1:], model.scale())
                bce = F.binary_cross_entropy_with_logits(fp_logits.float(), bits.view(len(sel), 1 + n_neg, -1)[:, 0])
                loss = nce + aux_weight * bce
            opt.zero_grad(set_to_none=True)
            scaler.scale(loss).backward()
            scaler.unscale_(opt)
            nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            scaler.step(opt)
            scaler.update()
            if sched.last_epoch < sched.total_steps - 1:
                sched.step()
            tot["nce"] += float(nce.detach())
            tot["bce"] += float(bce.detach())
            tot["n"] += 1
            if (time.time() - start) / 60 > max_minutes:
                break
        row = {"epoch": ep + 1, "minutes": round((time.time() - start) / 60, 2),
               "train_nce": tot["nce"] / max(tot["n"], 1), "train_bce": tot["bce"] / max(tot["n"], 1),
               "scale": float(model.scale().detach())}
        if score_fn is not None:
            model.eval()
            row["val_mrr"] = float(score_fn(model))
        hist.append(row)
        log(row)
        crit = row.get("val_mrr", -row["train_nce"])
        if crit > best:
            best, best_state = crit, {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
        if (time.time() - start) / 60 > max_minutes:
            log("retriever: stopped at the time budget")
            break
    model.load_state_dict(best_state)
    return model.eval(), pd.DataFrame(hist)


# ---------- inference ----------


@torch.no_grad()
def spectrum_vectors(model, mz, intensity, precursor, adduct, device, batch=2048):
    out = []
    for i in range(0, len(mz), batch):
        sl = slice(i, i + batch)
        out.append(model.spectra(torch.as_tensor(mz[sl], device=device),
                                 torch.as_tensor(intensity[sl], device=device),
                                 torch.as_tensor(precursor[sl], dtype=torch.float32, device=device),
                                 torch.as_tensor(adduct[sl], device=device)).float().cpu())
    return torch.cat(out) if out else torch.zeros(0, model.spectra.norm.normalized_shape[0])


@torch.no_grad()
def query_vectors(model, sets, H, device, mode="attention", chunk=1024):
    """(n_sets, E) numpy, one normalised query vector per molecule, plus the attention weights."""
    out, atts = [], []
    for s in range(0, len(sets), chunk):
        h, mask = pad_sets(sets[s:s + chunk], H)
        q, _, a = model.query(h.to(device), mask.to(device), mode)
        out.append(q.cpu().numpy())
        if a is not None:
            atts += [a[i, :len(x)].cpu().numpy() for i, x in enumerate(sets[s:s + chunk])]
    return np.concatenate(out), atts


@torch.no_grad()
def molecule_vectors(model, pool_fp, indices, device, chunk=8192):
    """{pool index: vector} for the given indices (deduplicated)."""
    uniq = np.unique(np.asarray(indices, dtype=np.int64))
    vecs = []
    for s in range(0, len(uniq), chunk):
        bits = _unpack(torch.as_tensor(pool_fp[uniq[s:s + chunk]], device=device))
        vecs.append(model.molecules(bits).cpu().numpy())
    mat = np.concatenate(vecs) if vecs else np.zeros((0, model.m_proj.out_features), np.float32)
    return uniq, mat


def window_scores(model, pool_fp, pool_mass, masses, q, device, ppm=10.0):
    """{molecule: (window indices, cosine score)} in the same form as ``full.window_similarity``."""
    lo, hi = window_bounds(pool_mass, masses.to_numpy(), ppm)
    uniq, mat = molecule_vectors(model, pool_fp, np.concatenate([np.arange(a, b) for a, b in zip(lo, hi,
                                                                                               strict=True)]
                                                                 + [np.zeros(0, np.int64)]), device)
    out = {}
    for (mol, _), a, b, qi in zip(masses.items(), lo, hi, q, strict=True):
        idx = np.arange(a, b)
        rows = np.searchsorted(uniq, idx)
        out[mol] = (idx, (mat[rows] @ qi).astype(np.float32) if len(idx) else np.zeros(0, np.float32))
    return out


def fuse_windows(a: dict, b: dict, wb: float = 1.0) -> dict:
    """Per-window z(a) + wb * z(b) of two score dicts over the same windows."""
    def z(x):
        return (x - x.mean()) / (x.std() + 1e-6) if len(x) else x
    return {m: (a[m][0], z(a[m][1]) + wb * z(b[m][1])) for m in a}
```

```python
# ---- Member 3's split, candidate pool (+ optional extra candidates), fingerprints ----
meta = cp.load_meta(DATA)
queries, lib_mask = cp.make_split(meta, max_query_spectra=16)
queries = cp.sample_molecules(queries, N_VAL)
truth = cp.truth_table(queries).reset_index(drop=True)
assert not lib_mask[queries.row_id.to_numpy()].any()

cand_files = find_files(CAND_ROOT)
print("candidate files:", {k: [os.path.basename(f) for f in v] for k, v in cand_files.items()})
if not cand_files:
    print("WARNING: no COCONUT / PubChemLite attached, pool = training structures only (v1 behaviour)")
elif not HAS_RDKIT:
    raise RuntimeError("COCONUT/PubChemLite attached but RDKit is missing: attach the rdkit_cp313 dataset.")
pubchem_files = find_pubchem(CAND_ROOT)
extra_smiles = {}
if pubchem_files and HAS_RDKIT and USE_PUBCHEM:
    test_meta, _ = cp.load_test(DATA)  # the test set Kaggle provides at this run (hidden one when scoring)
    query_masses = np.r_[queries.groupby("molecule_id").neutral_mass.apply(consensus_mass).to_numpy(),
                         test_meta.groupby("molecule_id").neutral_mass.apply(consensus_mass).to_numpy()]
    extra_smiles["pubchem"] = pubchem_in_windows(pubchem_files["mass"], pubchem_files["smiles"], query_masses, PPM,
                                                 cap=PUBCHEM_CAP, log=stamp)
else:
    print("PubChem off (POOL_USE_PUBCHEM=0, or files / RDKit missing)")
extra, extra_fp = (load_candidates(cand_files, n_bits=cp.FP_BITS, log=stamp, smiles_by_source=extra_smiles)
                   if (cand_files or extra_smiles) else (None, None))
pool = cp.build_pool(meta, lib_mask, extra_candidates=extra)
def pool_fingerprints(pool):
    """Morgan fingerprints for the pool: from an attached pool_fp_*.npz (keys + fp), RDKit for the rest."""
    fp = np.zeros((len(pool), cp.FP_BITS // 8), dtype=np.uint8)
    todo = np.ones(len(pool), dtype=bool)
    if extra is not None:  # external rows (empty formula) take the fingerprint computed while loading
        is_ext = (pool.formula == "").to_numpy()
        pos = dict(zip(extra.key, range(len(extra)), strict=True))
        hit = np.array([pos.get(k, -1) if e else -1 for k, e in zip(pool.key, is_ext, strict=True)], dtype=np.int64)
        fp[hit >= 0] = extra_fp[hit[hit >= 0]]
        todo &= hit < 0
        print(f"fingerprints from external sets: {(hit >= 0).sum():,}")
    for f in glob.glob("/kaggle/input/**/pool_fp_*.npz", recursive=True) + glob.glob(E("POOL_FP_CACHE", "") or "/none"):
        z = np.load(f, allow_pickle=True)
        if z["fp"].shape[1] != fp.shape[1]:
            continue
        cached = dict(zip(z["keys"].tolist(), range(len(z["keys"])), strict=True))
        hit = np.array([cached.get(k, -1) for k in pool.key], dtype=np.int64)
        fp[hit >= 0] = z["fp"][hit[hit >= 0]]
        todo &= hit < 0
        print(f"fingerprints from {f}: {(hit >= 0).sum():,} of {len(pool):,}")
    if todo.any():
        if not HAS_RDKIT:
            raise RuntimeError(f"{todo.sum():,} pool structures have no cached fingerprint and RDKit is missing. "
                               "Attach the pool_fp_2048.npz dataset (or an rdkit wheel).")
        fp[todo] = cp.morgan_packed(pool.smiles[todo].tolist())
    return fp

pool_fp = pool_fingerprints(pool)
key_index = {k: i for i, k in enumerate(pool.key.to_numpy())}
n_train_structures = meta.inchikey14.nunique()
print(f"pool {len(pool):,} structures ({len(pool) - n_train_structures:+,} beyond the training set) | "
      f"val {len(truth):,} molecules (class 1: {(truth.cls == 1).sum()}, class 2: {(truth.cls == 2).sum()})")
ext_keys = set(extra.key) if extra is not None else set()
np_keys = set(meta.loc[meta.ingest_lib == "enveda-np-examples", "inchikey14"])
cand_cov = {"external_structures_added": int((pool.formula == "").sum()),
            "np_examples_in_external": round(len(np_keys & ext_keys) / max(len(np_keys), 1), 3),
            "val_answers_in_external": round(float(truth.true_key.isin(ext_keys).mean()), 3)}
if extra is not None:
    for s, g in extra.groupby("source"):
        cand_cov[f"np_examples_in_{s}"] = round(len(np_keys & set(g.key)) / max(len(np_keys), 1), 3)
        cand_cov[f"val_answers_in_{s}"] = round(float(truth.true_key.isin(set(g.key)).mean()), 3)
curated = set(extra.key[extra.source.isin(["coconut", "pubchemlite"])]) if extra is not None else set()
trusted = pool.key.isin(curated).to_numpy().astype(np.float32)
cand_cov["pool_share_curated"] = round(float(trusted.mean()), 3)
print("candidate coverage:", json.dumps(cand_cov))

# formulas for external rows (build_pool blanks them), PubChem-only rows, and the F / I flag
IS_EXT = (pool.formula == "").to_numpy()
if extra is not None:
    ext = extra.set_index("key")
    pool.loc[IS_EXT, "formula"] = pool.key[IS_EXT].map(ext.formula).fillna("").to_numpy()
    PC_ONLY = IS_EXT & pool.key.map(ext.source).eq("pubchem").to_numpy()
    del ext
else:
    PC_ONLY = np.zeros(len(pool), dtype=bool)
HAS_FI = pool.formula.str.contains(r"(?:F|I)(?![a-z])", regex=True).to_numpy()
np_in_pool = pool.key.isin(np_keys).to_numpy()
print(f"PubChem-only rows: {PC_ONLY.sum():,} | NP examples with F or I: {HAS_FI[np_in_pool].mean():.3f} "
      f"| all pool rows with F or I: {HAS_FI.mean():.3f}")
# synthetic-library penalty (test only). Natural products = COCONUT or an NP example.
#   tier A: enveda-180 structures that are not natural products; tier B: other training structures that are not
coconut_keys = set(extra.key[extra.source == "coconut"]) if extra is not None else set()
lib_str = meta.ingest_lib.astype(str)
e180 = set(meta.inchikey14[lib_str == "enveda-180"])
natural = coconut_keys | np_keys
TIER_A_KEYS = e180 - natural
TIER_B_KEYS = set(meta.inchikey14) - e180 - natural
SYN_A, SYN_B = pool.key.isin(TIER_A_KEYS).to_numpy(), pool.key.isin(TIER_B_KEYS).to_numpy()
print(f"enveda-180 structures: {len(e180):,} | tier A (enveda-180, not natural): {len(TIER_A_KEYS):,} "
      f"| tier B (other training, not natural): {len(TIER_B_KEYS):,} | validation answers in A: "
      f"{truth.true_key.isin(TIER_A_KEYS).mean():.3f}, in B: {truth.true_key.isin(TIER_B_KEYS).mean():.3f} "
      f"| NP examples in A: {len(np_keys & TIER_A_KEYS)}, in B: {len(np_keys & TIER_B_KEYS)}")
del lib_str, e180
import gc

extra = extra_fp = ext_keys = None  # several GB with PubChem; nothing below needs them
gc.collect()
stamp("split and pool ready")
```

```python
# ---- Spectra: training library rows, validation queries, test ----
train_rows = cp.select_train_rows(meta, lib_mask, per_structure=PER_STRUCTURE)
train_rows = train_rows[train_rows.inchikey14.isin(key_index)].reset_index(drop=True)
TRAIN_PATH = os.path.join(DATA, "train.parquet")
tr_mz, tr_in = cp.read_spectra(TRAIN_PATH, train_rows.row_id.to_numpy())
va_mz, va_in = cp.read_spectra(TRAIN_PATH, queries.row_id.to_numpy())
test, test_path = cp.load_test(DATA)
te_mz, te_in = cp.read_spectra(test_path, test.row_id.to_numpy())
data = {"mz": tr_mz, "intensity": tr_in, "precursor": train_rows.precursor_mz.to_numpy(np.float32),
        "adduct": cp.adduct_ids(train_rows.adduct.astype(str)),
        "target_index": train_rows.inchikey14.map(key_index).to_numpy(np.int64), "fp_packed": pool_fp}
val = {"pool": pool, "pool_fp": pool_fp, "queries": queries, "mz": va_mz, "intensity": va_in, "truth": truth}
print(f"train {len(train_rows):,} spectra | val {len(queries):,} | test {len(test):,} spectra, "
      f"{test.molecule_id.nunique()} molecules")
stamp("spectra read")
```

```python
# ---- Stage A: Member 3's ResidualMLP, data-parallel over both T4s ----
enc = cp.build_models()["residual_mlp"](**ENC_KW)
model = torch.nn.DataParallel(enc, device_ids=list(range(N_GPU))) if N_GPU > 1 else enc
model, enc_hist = cp.train_model(model, data, val, DEV, epochs=ENC_EPOCHS, batch=512 * max(N_GPU, 1),
                                 lr=3e-4 * max(N_GPU, 1) ** 0.5, max_minutes=ENC_MAX_MIN,
                                 out_path=f"{OUT}/m7_encoder.pt")
enc = (model.module if isinstance(model, torch.nn.DataParallel) else model).to(DEV).eval()
for p in enc.parameters():
    p.requires_grad_(False)
enc_hist.to_csv(f"{OUT}/m7_results/encoder_history.csv", index=False)
print(enc_hist.tail(3).to_string())
stamp("stage A done")
```

```python
# ---- Embeddings and sets ----
H_tr = embed_all(enc, tr_mz, tr_in, data["precursor"], data["adduct"], DEV)
H_va = embed_all(enc, va_mz, va_in, queries.precursor_mz.to_numpy(np.float32),
                 cp.adduct_ids(queries.adduct.astype(str)), DEV)
H_te = embed_all(enc, te_mz, te_in, test.precursor_mz.to_numpy(np.float32), cp.adduct_ids(test.adduct.astype(str)), DEV)
EMB = H_tr.shape[1]
groups = group_indices(train_rows.inchikey14.to_numpy().astype(str))
train_keys = sorted(groups)
target_fp = {k: pool_fp[key_index[k]] for k in train_keys}

def molecule_sets(frame, order):
    pos = frame.assign(pos=np.arange(len(frame))).groupby("molecule_id").pos.apply(np.array)
    return [pos[m] for m in order]
val_sets = molecule_sets(queries, truth.molecule_id)
test_mols = test.molecule_id.drop_duplicates().tolist()
test_sets = molecule_sets(test, test_mols)
val_bits = unpack_bits(torch.as_tensor(np.stack([pool_fp[key_index[k]] for k in truth.true_key])))
val_mass = queries.groupby("molecule_id").neutral_mass.apply(consensus_mass).reindex(truth.molecule_id)
test_mass = test.groupby("molecule_id").neutral_mass.apply(consensus_mass).reindex(test_mols)
old_test_mass = test.groupby("molecule_id").neutral_mass.median().reindex(test_mols)
moved = int((np.abs(old_test_mass - test_mass) / test_mass * 1e6 > PPM).sum())
print("test adducts:", test.adduct.value_counts().to_dict())
print(f"test molecules whose mass moved > {PPM} ppm vs the median: {moved} of {len(test_mols)}")
queries["neutral_mass"] = queries.molecule_id.map(val_mass)  # B2 takes the median of these: keep them equal
test["neutral_mass"] = test.molecule_id.map(test_mass)
if CLIP_ON:  # Architecture 3 reads peaks, not bins: keep the most intense CLIP_PEAKS per spectrum
    CLIP_SPEC = {"train": dict(zip(("mz", "intensity"), ct.top_peaks(tr_mz, tr_in, CLIP_PEAKS), strict=True),
                               precursor=data["precursor"], adduct=data["adduct"]),
                 "val": ct.top_peaks(va_mz, va_in, CLIP_PEAKS), "test": ct.top_peaks(te_mz, te_in, CLIP_PEAKS)}
del tr_mz, tr_in, data
stamp("embeddings done")

# ---- Fragment-formula explanation for every candidate in every molecule's window ----
pool_mass_arr, pool_formula, pool_key_arr = pool.mass.to_numpy(), pool.formula.to_numpy(), pool.key.to_numpy()
formula_cache = {}

def explain_windows(sets, frame, mzs, ins, masses):
    adducts, modes = frame.adduct.astype(str).to_numpy(), frame.ionization_mode.astype(str).to_numpy()
    out = {}
    for rows, (mol, mass) in zip(sets, masses.items(), strict=True):
        idx = cp.window_indices(pool_mass_arr, mass, PPM)
        peaks = molecule_peaks(rows, mzs, ins, adducts, modes)
        cache = formula_cache.setdefault(mol, {})
        out[mol] = np.nan_to_num(window_scores(pool_formula[idx], peaks, cache=cache), nan=0.5)
    return out

EXPL_VAL = explain_windows(val_sets, queries, va_mz, va_in, val_mass)
EXPL_TEST = explain_windows(test_sets, test, te_mz, te_in, test_mass)
true_expl, mean_expl = [], []
for mol, key, mass in zip(truth.molecule_id, truth.true_key, val_mass, strict=True):
    idx = cp.window_indices(pool_mass_arr, mass, PPM)
    hit = np.flatnonzero(pool_key_arr[idx] == key) if len(idx) else []
    if len(hit):
        true_expl.append(EXPL_VAL[mol][hit[0]])
        mean_expl.append(EXPL_VAL[mol].mean())
true_expl, mean_expl = np.array(true_expl), np.array(mean_expl)
print(f"explanation score: true answers median {np.median(true_expl):.2f} "
      f"(share < 0.3: {(true_expl < 0.3).mean():.3f}) vs window average {np.median(mean_expl):.2f}")
stamp("fragment explanation done")
```

```python
# ---- Stage B: mean (no training), mean_tuned and attention_head trained in parallel, one per GPU ----
pool_keys = pool.key.to_numpy()
heads, hists = {}, []

N_SEL = min(400, len(truth))  # checkpoint criterion on a fixed subset: ranking ~8,000 candidates per molecule is slow

def val_mrr(probs):
    """Balanced validation MRR@25 of pooled probabilities on the first N_SEL molecules (checkpoint criterion)."""
    rank = rank_pooled(cp, pool, pool_fp, val_mass.iloc[:N_SEL], probs[:N_SEL], PPM)
    return cp.evaluate(rank, truth.iloc[:N_SEL], pool_keys).loc["balanced", "MRR@25"]

def train_one(method, seed, device):
    head = make_pool(method, EMB, cp.FP_BITS, head=enc.head)
    head, hist = train_pool(head, H_tr, groups, train_keys, target_fp, device, H_va, val_sets, val_bits,
                            epochs=POOL_EPOCHS, seed=seed, log=lambda m: None, score_fn=val_mrr)
    heads[(method, seed)] = head
    hists.append(hist.assign(method=method, seed=seed))
    stamp(f"{method} seed {seed} done (best val MRR {hist.val_mrr.max():.4f}, epoch {hist.val_mrr.idxmax() + 1})")

jobs = [(m, s) for s in SEEDS for m in ("mean_tuned", "attention_head")]
for i in range(0, len(jobs), len(DEVS)):
    batch_jobs = jobs[i:i + len(DEVS)]
    threads = [threading.Thread(target=train_one, args=(m, s, DEVS[j])) for j, (m, s) in enumerate(batch_jobs)]
    [t.start() for t in threads]
    [t.join() for t in threads]
pd.concat(hists).to_csv(f"{OUT}/m7_results/pooling_history.csv", index=False)

@torch.no_grad()
def pooled_probs(method, seed, sets, H):
    if method == "mean":
        out = []
        for s in range(0, len(sets), 1024):
            h, mask = pad_sets(sets[s:s + 1024], H)
            p = torch.sigmoid(enc.head(h.to(DEV).float()).float())
            m = mask.to(DEV).unsqueeze(-1).float()
            out.append(((p * m).sum(1) / m.sum(1).clamp(min=1)).cpu())
        return torch.cat(out).numpy()
    head = heads[(method, seed)].to(DEV).eval()
    logits, _ = batched_forward(head, sets, H, DEV)
    return torch.sigmoid(logits).cpu().numpy()

pv, pt = {}, {}
for method in ["mean", "mean_tuned", "attention_head"]:
    for seed in ([0] if method == "mean" else SEEDS):
        name = method if len(SEEDS) == 1 or method == "mean" else f"{method}_s{seed}"
        pv[name] = pooled_probs(method, seed, val_sets, H_va)
        pt[name] = pooled_probs(method, seed, test_sets, H_te)

# ---- Tune beta (curated prior), w (fragment explanation) and PubChem on/off for the best method ----
pool_keys = pool.key.to_numpy()
curated_true = truth.true_key.isin(curated).to_numpy()

SIMS = {}  # (method, split) -> similarity per window, computed once and reused across the grid

def ranking(name, split, beta, w, use_pc, k=25):
    probs, masses, expl = (pv.get(name), val_mass, EXPL_VAL) if split == "val" else (pt.get(name), test_mass,
                                                                                       EXPL_TEST)
    if (name, split) not in SIMS:
        SIMS[(name, split)] = window_similarity(cp, pool, pool_fp, masses, probs, PPM)
    prior = beta * trusted
    if not use_pc:
        prior = prior - 1e3 * PC_ONLY
    if split == "test":  # F / I block: natural-product prior, applied to the test set only
        prior = prior - 1e3 * HAS_FI
    extra_fn = (lambda m, i: w * (expl[m] - 1.0)) if w else None
    return rank_from_similarity(cp, pool, SIMS[(name, split)], masses, k=k, prior=prior, extra=extra_fn)

def groups(rank):
    rr = cp.rank_table(rank, truth, pool_keys).rr.to_numpy()
    return float(rr[curated_true].mean()), float(rr[~curated_true].mean())

base = {n: float(cp.evaluate(ranking(n, "val", 0.0, 0.0, True), truth, pool_keys).loc["balanced", "MRR@25"])
        for n in pv}
tune_on = max(base, key=base.get)
print("plain ranking (beta 0, w 0, with PubChem):", base, "| tuning on", tune_on)
beta_table = []
for use_pc in (True, False):
    for b in BETAS:
        for w in WEIGHTS:
            a_mrr, b_mrr = groups(ranking(tune_on, "val", b, w, use_pc))
            est = COV_CURATED * a_mrr + (COV_TRAIN_ONLY + (COV_PUBCHEM_ONLY if use_pc else 0.0)) * b_mrr
            beta_table.append({"pubchem": use_pc, "beta": b, "w": w, "mrr_curated": round(a_mrr, 4),
                               "mrr_other": round(b_mrr, 4), "est_LB": round(est, 4)})
        stamp(f"pubchem {use_pc} beta {b}: best est LB {max(r['est_LB'] for r in beta_table[-len(WEIGHTS):]):.4f}")
beta_table = pd.DataFrame(beta_table).sort_values("est_LB", ascending=False).reset_index(drop=True)
beta_table.to_csv(f"{OUT}/m7_results/tuning_grid.csv", index=False)
print(beta_table.head(8).to_string(index=False))
BEST_CFG = beta_table.iloc[0].to_dict()
BETA, W_EXPL, USE_PC = float(BEST_CFG["beta"]), float(BEST_CFG["w"]), bool(BEST_CFG["pubchem"])
print(f"chosen: PubChem {USE_PC}, beta {BETA}, w {W_EXPL} | estimated LB {BEST_CFG['est_LB']:.3f} "
      f"(assumes coverage curated {COV_CURATED}, train-only {COV_TRAIN_ONLY}, PubChem-only {COV_PUBCHEM_ONLY})")
val_rank, test_rank = {}, {}
for name in pv:
    val_rank[name] = ranking(name, "val", BETA, W_EXPL, USE_PC)
    test_rank[name] = ranking(name, "test", BETA, W_EXPL, USE_PC, k=TEST_K)
stamp("stage B done")
```

```python
# ---- Architecture 3: contrastive peak transformer, trained on both T4s (InfoNCE with same-mass hard negatives) ----
import traceback

CLIP, CLIP_INFO = None, {}
try:
    if not CLIP_ON:
        raise RuntimeError("off (POOL_CLIP=0)")
    clip_groups = group_indices(train_rows.inchikey14.to_numpy().astype(str))  # `groups` is a function by now
    va_spec = dict(zip(("mz", "intensity"), CLIP_SPEC["val"], strict=True),
                   precursor=queries.precursor_mz.to_numpy(np.float32),
                   adduct=cp.adduct_ids(queries.adduct.astype(str)))
    te_spec = dict(zip(("mz", "intensity"), CLIP_SPEC["test"], strict=True),
                   precursor=test.precursor_mz.to_numpy(np.float32), adduct=cp.adduct_ids(test.adduct.astype(str)))
    n_sel = min(600, len(truth))
    sel_mass = val_mass.iloc[:n_sel]

    def clip_val(model):
        """Checkpoint criterion: balanced MRR@25 of the plain contrastive ranking on the first 600 molecules."""
        H = ct.spectrum_vectors(model, device=DEV, **va_spec)
        q, _ = ct.query_vectors(model, val_sets[:n_sel], H, DEV)
        sims = ct.window_scores(model, pool_fp, pool_mass_arr, sel_mass, q, DEV, PPM)
        rank = rank_from_similarity(cp, pool, sims, sel_mass)
        return float(cp.evaluate(rank, truth.iloc[:n_sel], pool_keys).loc["balanced", "MRR@25"])

    retr = ct.Retriever(**CLIP_KW, fp_bits=cp.FP_BITS, n_adducts=cp.N_ADDUCTS)
    n_params = sum(p.numel() for p in retr.parameters())
    stamp(f"contrastive transformer: {n_params / 1e6:.1f}M parameters, {len(train_keys):,} training molecules, "
          f"{CLIP_NEG} same-mass negatives each, {CLIP_PEAKS} peak tokens")
    retr, clip_hist = ct.train_retriever(retr, CLIP_SPEC["train"], clip_groups, train_keys, key_index, pool_fp,
                                         pool_mass_arr, DEV, epochs=CLIP_EPOCHS, batch=CLIP_BATCH, n_neg=CLIP_NEG,
                                         lr=3e-4 * max(N_GPU, 1) ** 0.5, ppm=PPM, max_minutes=CLIP_MAX_MIN,
                                         n_gpu=N_GPU, score_fn=clip_val, log=print)
    clip_hist.to_csv(f"{OUT}/m7_results/contrastive_history.csv", index=False)
    del CLIP_SPEC["train"], clip_groups
    gc.collect()
    H_va_c = ct.spectrum_vectors(retr, device=DEV, **va_spec)
    H_te_c = ct.spectrum_vectors(retr, device=DEV, **te_spec)
    Q = {}
    for mode, name in (("attention", "clip"), ("mean", "clip_mean")):
        qv, att = ct.query_vectors(retr, val_sets, H_va_c, DEV, mode)
        qt, _ = ct.query_vectors(retr, test_sets, H_te_c, DEV, mode)
        Q[name] = {"val": qv, "test": qt}
        SIMS[(name, "val")] = ct.window_scores(retr, pool_fp, pool_mass_arr, val_mass, qv, DEV, PPM)
        SIMS[(name, "test")] = ct.window_scores(retr, pool_fp, pool_mass_arr, test_mass, qt, DEV, PPM)
        if mode == "attention":
            multi = [a for a in att if len(a) >= 2]
            CLIP_INFO["attention_top_weight"] = round(float(np.mean([a.max() for a in multi])), 3) if multi else None
            CLIP_INFO["uniform_top_weight"] = round(float(np.mean([1 / len(a) for a in multi])), 3) if multi else None
    for split in ("val", "test"):
        ranking(tune_on, split, 0.0, 0.0, True)  # makes sure the MLP's window scores exist
        a, b = SIMS[(tune_on, split)], SIMS[("clip", split)]
        assert all(len(a[m][0]) == len(b[m][0]) for m in a), "window mismatch between first stages"
        SIMS[("fused", split)] = ct.fuse_windows(a, b)
    for name in ("clip", "clip_mean", "fused"):
        val_rank[name] = ranking(name, "val", BETA, W_EXPL, USE_PC)
        test_rank[name] = ranking(name, "test", BETA, W_EXPL, USE_PC, k=TEST_K)
    first = {n: float(cp.evaluate(val_rank[n], truth, pool_keys).loc["balanced", "MRR@25"])
             for n in (tune_on, "clip", "clip_mean", "fused")}
    CLIP_MODE = "clip" if first["clip"] >= first["clip_mean"] else "clip_mean"
    CLIP = {"model": retr, "q": Q[CLIP_MODE], "mode": CLIP_MODE}
    CLIP_INFO.update({"params_M": round(n_params / 1e6, 2), "epochs_run": int(len(clip_hist)),
                      "best_epoch": int(clip_hist.val_mrr.idxmax() + 1), "minutes": float(clip_hist.minutes.iloc[-1]),
                      "first_stage_val_MRR": {k: round(v, 4) for k, v in first.items()}, "features_from": CLIP_MODE})
    print(clip_hist.tail(3).to_string(index=False))
    print("first stage, validation MRR@25:", CLIP_INFO["first_stage_val_MRR"])
    print(f"attention pooling: weight of the top spectrum in sets of 2+ = {CLIP_INFO['attention_top_weight']} "
          f"(uniform = {CLIP_INFO['uniform_top_weight']})")
    stamp("contrastive transformer done")
except Exception as exc:  # Architecture 3 must not cost the run: the v7 pipeline continues without it
    if CLIP_ON:
        traceback.print_exc()
    CLIP_INFO = {"error": repr(exc)[:300]}
    print("CONTRASTIVE TRANSFORMER SKIPPED:", CLIP_INFO["error"])
```

```python
# ---- PubChem documentation prior for every validation and test window (used by the reranker and the channel) ----
import traceback

DOC, DOC_ERR = None, None
try:
    if not USE_DOCS:
        raise RuntimeError("off in v9 (POOL_USE_DOCS=0): v8 with it scored 0.276 vs v7's 0.281")
    docs, pcf = find_pubchem_docs(CAND_ROOT), find_pubchem(CAND_ROOT)
    needs = (("pubchem-2026-11 (CID-Mass, CID-SMILES)", pcf), ("pubchem-docs-2026-11 (CID-SID, CID-PMID)", docs),
             ("RDKit", HAS_RDKIT))
    missing = [name for name, ok in needs if not ok]
    if missing:
        raise RuntimeError("missing input: " + ", ".join(missing))
    n_val_w = len(truth)  # windows: validation molecules in truth order, then test molecules in test_mols order
    masses_all = np.r_[val_mass.to_numpy(), test_mass.to_numpy()]
    wc, wm = window_compounds(pcf["mass"], masses_all, PPM)
    stamp(f"documentation: {len(wc):,} neutral organic PubChem compounds in {len(masses_all)} windows")
    pop_w = documentation_prior(count_rows_per_cid(docs["sid"], wc), count_rows_per_cid(docs["pmid"], wc))
    stamp(f"documentation: counts done (share with pop > 0: {(pop_w > 0).mean():.3f})")
    lists, union = shortlist_windows(wc, wm, pop_w, masses_all, PPM, CH_DEPTH)
    stamp(f"documentation: {len(union):,} best-documented CIDs (top {CH_DEPTH} per window)")
    CH, CH_FP = convert_cids(smiles_for_cids(pcf["smiles"], union), n_bits=cp.FP_BITS)
    srt = np.argsort(wc)
    CH["pop"] = pop_w[srt][np.searchsorted(wc[srt], CH.cid.to_numpy())]
    n_window_cids = int(len(wc))
    del wc, wm, pop_w, srt
    gc.collect()
    cid_key = dict(zip(CH.cid.tolist(), CH.key.tolist(), strict=True))
    cid_pop = dict(zip(CH.cid.tolist(), CH["pop"].tolist(), strict=True))

    def window_key_pop(j):
        out = {}
        for c in lists[j].tolist():
            k = cid_key.get(c)
            if k is not None and cid_pop[c] > out.get(k, -1.0):
                out[k] = cid_pop[c]
        return out

    DOC = {"lists": lists, "CH": CH, "CH_FP": CH_FP, "n_val": n_val_w, "key_pop": window_key_pop,
           "n_window_cids": n_window_cids}
    ans_pop = [window_key_pop(j).get(k) for j, k in enumerate(truth.true_key)]
    print(f"validation answers among their window's top {CH_DEPTH} documented: "
          f"{np.mean([p is not None for p in ans_pop]):.3f}")
    stamp(f"documentation prior ready: {len(CH):,} structures")
except Exception as exc:  # without it the reranker runs as in v7 and the channel is off
    if USE_DOCS:
        traceback.print_exc()
    DOC_ERR = repr(exc)[:300]
    print("documentation prior:", DOC_ERR)
```

```python
# ---- Stage 2: line-up reranker (out-of-fold on validation, then trained on all of it for the test) ----
import traceback

from common import rerank as rr

RERANK = {}
try:
    K = RERANK_K
    # first stage for the line-ups: the MLP, the contrastive transformer or both fused, whichever ranks best
    first_opts = {n: float(cp.evaluate(val_rank[n], truth, pool_keys).loc["balanced", "MRR@25"])
                  for n in (tune_on, "clip", "fused") if n in val_rank}
    first_name = max(first_opts, key=first_opts.get)
    print("line-up first stage:", first_name, {k: round(v, 4) for k, v in first_opts.items()})
    pool_smiles = pool.smiles.to_numpy()

    def first_lists(split, k):
        return ranking(first_name, split, BETA, W_EXPL, USE_PC, k=k)

    def first_scores(split, mol, cand):
        idx, tan = SIMS[(first_name, split)][mol]
        expl = (EXPL_VAL if split == "val" else EXPL_TEST)[mol]
        out = np.full(len(cand), -1.0)
        if len(idx):
            pos = np.asarray(cand) - idx[0]
            ok = (pos >= 0) & (pos < len(idx))
            p = pos[ok]
            out[ok] = tan[p] + BETA * trusted[np.asarray(cand)[ok]] + W_EXPL * (expl[p] - 1.0)
        return out

    cand_val = first_lists("val", K)
    cand_test = first_lists("test", max(K, TEST_K))
    stamp(f"reranker: line-ups of {K} for {len(cand_val)} validation and {len(cand_test)} test molecules")
    if CLIP is not None:  # contrastive score of every line-up candidate
        CU, CM = ct.molecule_vectors(CLIP["model"], pool_fp, np.concatenate(
            [np.asarray(c) for c in list(cand_val.values()) + list(cand_test.values())]), DEV)

        def clip_sim(split, i, cand):
            return CM[np.searchsorted(CU, cand)] @ CLIP["q"][split][i]

    # library score of every candidate (validation library excludes held-out spectra; test uses all)
    lib_val = cp.SpectralLibrary(DATA, meta, lib_mask, pool, val_mass.to_numpy(), ppm=PPM)
    LIB_VAL = candidate_lib_scores(cp, pool, lib_val, queries, va_mz, va_in, cand_val)
    del lib_val
    lib_test = cp.SpectralLibrary(DATA, meta, np.ones(len(meta), dtype=bool), pool, test_mass.to_numpy(), ppm=PPM)
    LIB_TEST = candidate_lib_scores(cp, pool, lib_test, test, te_mz, te_in, cand_test)
    del lib_test
    stamp("reranker: library scores done")

    all_smiles = [pool_smiles[i] for c in list(cand_val.values()) + list(cand_test.values()) for i in c]
    STRUCT = rr.precompute_structures(all_smiles)
    stamp(f"reranker: bond-cleavage tables for {len(STRUCT):,} structures")

    def lineups(split):
        sets_, frame, mzs, ins, masses, cands, libs, probs = (
            (val_sets, queries, va_mz, va_in, val_mass, cand_val, LIB_VAL, pv) if split == "val"
            else (test_sets, test, te_mz, te_in, test_mass, cand_test, LIB_TEST, pt))
        adducts, modes = frame.adduct.astype(str).to_numpy(), frame.ionization_mode.astype(str).to_numpy()
        precs = frame.precursor_mz.to_numpy(float)
        tables = []
        for i, (rows, (mol, mass)) in enumerate(zip(sets_, masses.items(), strict=True)):
            cand = np.asarray(cands[mol])
            ev = rr.spectrum_evidence(rows, mzs, ins, precs, adducts, modes)
            t = rr.lineup_features(cand, first_scores(split, mol, cand), {m: probs[m][i] for m in rr.METHODS},
                                   pool_fp, pool_smiles, pool_formula, pool_mass_arr, mass, ev, STRUCT, libs[mol])
            if DOC is not None:
                kp = DOC["key_pop"](i if split == "val" else DOC["n_val"] + i)
                t = pd.concat([t.reset_index(drop=True), rr.doc_features(pool_keys[cand], pool_formula[cand], kp)],
                              axis=1)
            if CLIP is not None:
                cf = rr.clip_features(clip_sim(split, i, cand), pool_formula[cand])
                t = pd.concat([t.reset_index(drop=True), cf], axis=1)
            t.insert(0, "cand", cand)
            t.insert(0, "group", mol)
            tables.append(t)
        return pd.concat(tables, ignore_index=True)

    TAB_VAL, TAB_TEST = lineups("val"), lineups("test")
    TAB_VAL["label"] = (pool_keys[TAB_VAL.cand.to_numpy()] == TAB_VAL.group.map(
        dict(zip(truth.molecule_id, truth.true_key, strict=True))).to_numpy()).astype(int)
    in_k = TAB_VAL.groupby("group").label.max()
    print(f"answer inside the top {K}: {in_k.mean():.3f} of validation molecules")
    stamp("reranker: features done")

    def lists_from(table, score, k=25):
        t = table.assign(s=score).sort_values(["group", "s"], ascending=[True, False], kind="stable")
        return {m: g.cand.tolist()[:k] for m, g in t.groupby("group", sort=False)}

    def mrr_of(lists):
        return float(cp.evaluate(lists, truth, pool_keys).loc["balanced", "MRR@25"])

    def np_of(lists):
        return float(cp.evaluate(lists, truth, pool_keys).loc["np proxy", "MRR@25"])

    # ablation: which evidence earns what (each set cross-fitted on the same folds)
    sets = {"first stage only": rr.GROUPS["base"],
            "+ fingerprint likelihood": rr.GROUPS["base"] + rr.GROUPS["fp"],
            "+ exact masses & losses": rr.GROUPS["base"] + rr.GROUPS["fp"] + rr.GROUPS["ms"],
            "+ bond cleavage": rr.GROUPS["base"] + rr.GROUPS["fp"] + rr.GROUPS["ms"] + rr.GROUPS["bond"],
            "+ library (all)": rr.ALL_FEATURES}
    if DOC is not None:
        sets["+ documentation prior"] = rr.ALL_FEATURES + rr.DOC_FEATURES
    if CLIP is not None:
        sets["+ contrastive transformer"] = rr.ALL_FEATURES + rr.CLIP_FEATURES
    ablation, oofs = [], {}
    stage1 = {m: list(c)[:25] for m, c in cand_val.items()}
    stage1_mrr = mrr_of(stage1)
    for name, feats in sets.items():
        oofs[name] = rr.cross_fit(TAB_VAL, feats)
        lst = lists_from(TAB_VAL, oofs[name])
        row = {"features": name, "balanced_MRR": round(mrr_of(lst), 4), "np_proxy_MRR": round(np_of(lst), 4)}
        ablation.append(row)
        stamp(f"reranker ablation: {name}: {row['balanced_MRR']:.4f} (NP examples {row['np_proxy_MRR']:.4f})")
    ablation.insert(0, {"features": "no reranker (stage 1)", "balanced_MRR": round(stage1_mrr, 4),
                        "np_proxy_MRR": round(np_of(stage1), 4)})
    ABLATION = pd.DataFrame(ablation)
    print(ABLATION.to_string(index=False))
    abl = ABLATION.set_index("features")
    # validation answers are library standards (well documented), so the prior must also lift the NP examples
    eligible = [n for n in sets if n != "+ documentation prior"
                or abl.np_proxy_MRR[n] > abl.np_proxy_MRR["+ library (all)"]]
    if DOC is not None and "+ documentation prior" not in eligible:
        print("documentation prior not used: it does not lift the NP examples")
    BEST_SET = max(eligible, key=lambda n: abl.balanced_MRR[n])
    if ABLATION.set_index("features").balanced_MRR[BEST_SET] <= stage1_mrr:
        raise RuntimeError(f"no reranker beats stage 1 ({stage1_mrr:.4f}); keeping the stage-1 order")
    FEATS, OOF = sets[BEST_SET], oofs[BEST_SET]
    print(f"reranker uses: {BEST_SET} ({len(FEATS)} features)")

    # curated-source bonus on the reranker's (list-standardised) score, tuned with the v5 estimate
    def zscore(table, s):
        g = table.assign(s=s).groupby("group").s
        return ((table.assign(s=s).s - g.transform("mean")) / (g.transform("std") + 1e-6)).to_numpy()

    Z_VAL = zscore(TAB_VAL, OOF)
    cur_val = trusted[TAB_VAL.cand.to_numpy()]
    grid = []
    for b2 in (0.0, 0.25, 0.5, 1.0, 2.0):
        a_mrr, b_mrr = groups(lists_from(TAB_VAL, Z_VAL + b2 * cur_val))
        est = COV_CURATED * a_mrr + (COV_TRAIN_ONLY + (COV_PUBCHEM_ONLY if USE_PC else 0.0)) * b_mrr
        grid.append({"bonus": b2, "mrr_curated": round(a_mrr, 4), "mrr_other": round(b_mrr, 4),
                     "est_LB": round(est, 4)})
    RR_GRID = pd.DataFrame(grid).sort_values("est_LB", ascending=False).reset_index(drop=True)
    print(RR_GRID.to_string(index=False))
    B2_BONUS = float(RR_GRID.bonus[0])
    val_rank["reranked"] = lists_from(TAB_VAL, Z_VAL + B2_BONUS * cur_val)

    # final model on every validation line-up -> test
    FINAL = rr.fit_bag(TAB_VAL, FEATS, RR_SEEDS)
    s_test = zscore(TAB_TEST, rr.predict(FINAL, TAB_TEST, FEATS))
    s_test = s_test + B2_BONUS * trusted[TAB_TEST.cand.to_numpy()] - 1e3 * HAS_FI[TAB_TEST.cand.to_numpy()]
    test_rank["reranked"] = lists_from(TAB_TEST, s_test, k=TEST_K)
    imp = pd.Series(np.mean([m.feature_importance("gain") for m in FINAL], axis=0), index=FEATS).sort_values(
        ascending=False)
    print("top features (gain):", imp.head(12).round(0).to_dict())

    # where the remaining loss is: every validation molecule after reranking
    fam = dict(zip(truth.molecule_id, truth.true_key, strict=True))
    rows_ = []
    for mol, lst in val_rank["reranked"].items():
        keys_ = [pool_keys[i] for i in lst]
        ans = fam[mol]
        r = keys_.index(ans) + 1 if ans in keys_ else 0
        ans_idx = TAB_VAL.cand[(TAB_VAL.group == mol) & (TAB_VAL.label == 1)]
        f_ans = pool_formula[int(ans_idx.iloc[0])] if len(ans_idx) else None
        f_top = pool_formula[lst[0]] if lst else None
        cat = ("rank 1" if r == 1 else "answer not in line-up" if not in_k.get(mol, 0) else
               "beaten by same-formula isomer" if f_ans == f_top else "beaten by a different formula")
        rows_.append({"molecule_id": mol, "rank": r, "category": cat, "rr": 1.0 / r if r else 0.0})
    LOSS = pd.DataFrame(rows_)
    LOSS.to_csv(f"{OUT}/m7_results/val_rerank_losses.csv", index=False)
    summary_loss = LOSS.groupby("category").agg(molecules=("rr", "size"), mrr_lost=("rr", lambda x: (1 - x).sum()))
    summary_loss["share_of_loss"] = (summary_loss.mrr_lost / summary_loss.mrr_lost.sum()).round(3)
    print(summary_loss.sort_values("mrr_lost", ascending=False).to_string())
    ABLATION.to_csv(f"{OUT}/m7_results/rerank_ablation.csv", index=False)
    RR_GRID.to_csv(f"{OUT}/m7_results/rerank_bonus_grid.csv", index=False)
    imp.to_csv(f"{OUT}/m7_results/rerank_feature_gain.csv")
    RERANK = {"K": K, "seeds": list(RR_SEEDS), "documentation": DOC is not None, "first_stage": first_name,
              "contrastive_features": CLIP is not None,
              "features": BEST_SET, "answer_in_lineup": round(float(in_k.mean()), 4), "bonus": B2_BONUS,
              "ablation": ABLATION.to_dict("records"), "loss": summary_loss.reset_index().to_dict("records")}
    stamp("reranker done")
except Exception as exc:  # a reranker failure must not cost the run: fall back to the stage-1 lists
    traceback.print_exc()
    RERANK = {"error": repr(exc)[:300]}
    if SUBMIT == "reranked":
        SUBMIT = tune_on
    print(f"RERANKER SKIPPED, submission falls back to the stage-1 lists of {tune_on}:", RERANK)
```

```python
# ---- PubChem channel: documentation prior, then a gated merge into the main (reranked) lists ----
import traceback

CHANNEL, POOL_OUT_FRAME = {}, pool
try:
    if DOC is None:
        raise RuntimeError(f"documentation prior not available ({DOC_ERR})")
    base_name = "reranked" if "reranked" in val_rank else tune_on
    rng_ = np.random.default_rng(5)
    np_mols = truth.molecule_id[truth.query_lib == "enveda-np-examples"].tolist()
    others = truth.molecule_id[truth.query_lib != "enveda-np-examples"].tolist()
    ch_val = np_mols + list(rng_.choice(others, size=min(CH_OTHERS, len(others)), replace=False))
    CH, CH_FP, lists = DOC["CH"], DOC["CH_FP"], DOC["lists"]
    CH = CH.assign(row=np.arange(len(CH))).sort_values("pop", ascending=False, kind="stable")
    CHK = CH.drop_duplicates("key").reset_index(drop=True)  # one row per structure, its best-documented CID
    CHK_FP = CH_FP[CHK.row.to_numpy()]
    key_row = dict(zip(CHK.key, range(len(CHK)), strict=True))
    cid_row = {c: key_row[k] for c, k in zip(CH.cid, CH.key, strict=True)}
    OFFSET = len(pool)
    POOL_EXT = pd.concat([pool[["key", "smiles"]], CHK[["key", "smiles"]]], ignore_index=True)
    ext_keys = POOL_EXT.key.to_numpy()
    in_pool = set(pool_keys)
    stamp(f"channel: {len(CHK):,} shortlisted structures")

    def channel(j, probs, lam, answer=None):
        rows = sorted({cid_row[c] for c in lists[j] if c in cid_row})
        rows = np.array([r for r in rows if CHK.key[r] not in in_pool or CHK.key[r] == answer], dtype=np.int64)
        if len(rows) == 0:
            return [], np.array([])
        b = np.unpackbits(CHK_FP[rows], axis=1).astype(np.float32)
        ov = b @ probs
        tan = ov / (probs.sum() + b.sum(1) - ov + 1e-8)
        s = (tan - tan.mean()) / (tan.std() + 1e-6) + lam * CHK["pop"].to_numpy()[rows]
        o = np.argsort(-s, kind="stable")[:25]
        return [OFFSET + int(r) for r in rows[o]], s[o]

    pos_val = {m: i for i, m in enumerate(truth.molecule_id)}
    ans = dict(zip(truth.molecule_id, truth.true_key, strict=True))

    def rr(lst, key):
        ks = [ext_keys[i] for i in lst[:25]]
        return 1.0 / (ks.index(key) + 1) if key in ks else 0.0

    # lambda: the NP examples with their answer removed from the main pool
    lam_table = []
    for lam in LAMS:
        v = [rr(channel(pos_val[m], pv[tune_on][pos_val[m]], lam, ans[m])[0], ans[m]) for m in np_mols]
        lam_table.append({"lambda": lam, "channel_MRR_np_examples": round(float(np.mean(v)), 4)})
    LAM_TABLE = pd.DataFrame(lam_table)
    print(LAM_TABLE.to_string(index=False))
    LAM = float(LAM_TABLE.sort_values("channel_MRR_np_examples", ascending=False)["lambda"].iloc[0])

    # merge policy: cost to answers in the main lists vs gain for answers only PubChem has
    chan_ex = {m: channel(pos_val[m], pv[tune_on][pos_val[m]], LAM) for m in ch_val}
    chan_np = {m: channel(pos_val[m], pv[tune_on][pos_val[m]], LAM, ans[m]) for m in np_mols}
    s0 = np.array([c[1][0] for c in chan_ex.values() if len(c[1])])
    taus = [None] + [float(np.quantile(s0, q)) for q in (0.5, 0.75, 0.9)] if len(s0) else [None]
    slot_sets = {"none": (), "2-10 even": (2, 4, 6, 8, 10), "4-20 by 4": (4, 8, 12, 16, 20),
                 "10-24 even": (10, 12, 14, 16, 18, 20, 22, 24)}
    policies = []
    for sname, slots in slot_sets.items():
        for tau in taus:
            def merged(m, c, drop=None, slots=slots, tau=tau):
                main = [i for i in val_rank[base_name][m] if drop is None or pool_keys[i] != drop]
                ovr = bool(tau is not None and len(c[1]) and c[1][0] >= tau)
                return merge_channel(main, c[0], slots, override=ovr)
            mrr_main = float(np.mean([rr(merged(m, chan_ex[m]), ans[m]) for m in ch_val]))
            mrr_pc = float(np.mean([rr(merged(m, chan_np[m], drop=ans[m]), ans[m]) for m in np_mols]))
            row = {"slots": sname, "rank1_if_S_over": "never" if tau is None else round(tau, 2),
                   "mrr_main_answers": round(mrr_main, 4), "mrr_pubchem_only": round(mrr_pc, 4)}
            for c_ in (0.10, 0.25, 0.40):
                row[f"est_LB_c{c_:.2f}"] = round(COV_MAIN * mrr_main + c_ * mrr_pc, 4)
            row["est_LB"] = round(COV_MAIN * mrr_main + COV_PC * mrr_pc, 4)
            policies.append(row)
    POLICY = pd.DataFrame(policies).sort_values("est_LB", ascending=False).reset_index(drop=True)
    print(POLICY.head(10).to_string(index=False))
    best_policy = POLICY.iloc[0].to_dict()
    slots, tau = slot_sets[best_policy["slots"]], best_policy["rank1_if_S_over"]
    tau = None if tau == "never" else float(tau)
    if not slots and tau is None:
        raise RuntimeError("best policy is to add nothing from PubChem; keeping the main lists")

    test_pos = {m: i for i, m in enumerate(test_mols)}
    merged_test = {}
    for m in test_mols:
        c = channel(DOC["n_val"] + test_pos[m], pt[tune_on][test_pos[m]], LAM)
        ovr = bool(tau is not None and len(c[1]) and c[1][0] >= tau)
        merged_test[m] = merge_channel(test_rank[base_name][m], c[0], slots, override=ovr, n=TEST_K)
    test_rank["reranked+pubchem"] = merged_test
    POOL_OUT_FRAME = POOL_EXT
    if SUBMIT in ("reranked", "auto"):
        SUBMIT = "reranked+pubchem"
    LAM_TABLE.to_csv(f"{OUT}/m7_results/channel_lambda.csv", index=False)
    POLICY.to_csv(f"{OUT}/m7_results/channel_merge_policies.csv", index=False)
    CHANNEL = {"lambda": LAM, "lambda_table": LAM_TABLE.to_dict("records"), "policy": best_policy,
               "shortlisted": int(len(CHK)), "window_compounds": DOC["n_window_cids"]}
    stamp(f"channel done: lambda {LAM}, slots {best_policy['slots']}, rank-1 rule {best_policy['rank1_if_S_over']}")
except Exception as exc:  # the channel must not cost the run: keep the main lists
    if DOC is not None:
        traceback.print_exc()
    CHANNEL = {"error": repr(exc)[:300]}
    print("PubChem channel:", CHANNEL["error"])
```

```python
# ---- Synthetic-library penalty (test only): tier A to the end of the top 25, tier B after the rest on ranks 2-25 ----
SYN_INFO = {"on": SYN_DEMOTE, "tier_A": int(SYN_A.sum()), "tier_B": int(SYN_B.sum())}
base_name = SUBMIT if SUBMIT in test_rank else None
if SYN_DEMOTE and (SYN_A.any() or SYN_B.any()) and base_name is not None:
    def tier(i):  # 0 unflagged, 1 tier B, 2 tier A; channel rows (index >= len(pool)) are never flagged
        return 0 if i >= len(SYN_A) else 2 if SYN_A[i] else 1 if SYN_B[i] else 0

    def demote_one(lst, k=25):
        top, rest = list(lst[:k]), list(lst[k:])
        head = top[:1] if top and tier(top[0]) < 2 else []  # rank 1 moves only for tier A
        body = top[len(head):]
        return head + sorted(body, key=tier) + rest  # stable: order kept within each tier

    def demote(lists):
        return {m: demote_one(lst) for m, lst in lists.items()}

    name = base_name + "+syn"
    test_rank[name] = demote(test_rank[base_name])
    top1_changed = sum(test_rank[name][m][0] != test_rank[base_name][m][0] for m in test_mols)
    tiers = np.array([tier(i) for m in test_mols for i in test_rank[base_name][m][:25]])
    top1_tiers = np.array([tier(test_rank[base_name][m][0]) for m in test_mols])
    SYN_INFO.update({"applied_to": base_name, "test_top1_changed": int(top1_changed),
                     "test_top1_tier_A": int((top1_tiers == 2).sum()), "test_top1_tier_B": int((top1_tiers == 1).sum()),
                     "test_top25_share_A": round(float((tiers == 2).mean()), 4),
                     "test_top25_share_B": round(float((tiers == 1).mean()), 4)})
    if base_name in val_rank:  # what it would do on validation, whose answers include these structures
        val_rank[name] = demote(val_rank[base_name])
        t_before, t_after = (cp.evaluate(val_rank[n], truth, pool_keys) for n in (base_name, name))
        SYN_INFO["val_what_if"] = {g: [round(float(tb.loc[g, "MRR@25"]), 4) for tb in (t_before, t_after)]
                                   for g in ("balanced", "class 1", "class 2", "np proxy")}
    SUBMIT = name
    print(f"synthetic penalty on {base_name}: test top-1 was tier A for {SYN_INFO['test_top1_tier_A']}, tier B for "
          f"{SYN_INFO['test_top1_tier_B']} of {len(test_mols)}; top-1 changed for {top1_changed} | share of test "
          f"top-25 slots: tier A {SYN_INFO['test_top25_share_A']:.3f}, tier B {SYN_INFO['test_top25_share_B']:.3f}")
    print("validation what-if [before, after] (validation answers include these structures, so it must drop "
          "except on the NP examples):", SYN_INFO.get("val_what_if"))
else:
    print("synthetic penalty off:", SYN_INFO)
```

```python
# ---- Baselines B1 and B2, plain fusion, and the gated B2 fusion ----
val_rank["B1"] = {m: cp.rank_b1(pool, val_mass[m], PPM) for m in truth.molecule_id}
test_rank["B1"] = {m: cp.rank_b1(pool, test_mass[m], PPM) for m in test_mols}
if RUN_B2:
    lib = cp.SpectralLibrary(DATA, meta, lib_mask, pool, val_mass.to_numpy(), ppm=PPM)
    val_rank["B2"] = cp.run_b2(pool, lib, queries, va_mz, va_in, ppm=PPM, pooling="mean")
    val_b2_score = b2_top_scores(cp, pool, lib, queries, va_mz, va_in, val_rank["B2"])
    full_lib = np.ones(len(meta), dtype=bool)  # test pass: every training spectrum is a valid reference
    lib = cp.SpectralLibrary(DATA, meta, full_lib, pool, test_mass.to_numpy(), ppm=PPM)
    test_rank["B2"] = cp.run_b2(pool, lib, test, te_mz, te_in, ppm=PPM, pooling="mean")
    test_b2_score = b2_top_scores(cp, pool, lib, test, te_mz, te_in, test_rank["B2"])
    del lib
    stamp("B2 done")

def score(rankings):
    return cp.evaluate(rankings, truth, pool_keys, ci=True)

tables = {name: score(r) for name, r in val_rank.items()}
honest = {name: float(t.loc["balanced", "MRR@25"]) for name, t in tables.items()}
fp_names = [n for n in val_rank if n not in ("B1", "B2") and not n.startswith("reranked")]
best_fp = max(fp_names, key=lambda n: honest[n])
gate_info = None
if RUN_B2:
    for w in (1.0, 2.0):
        name = f"fusion(B2 x{w})"
        val_rank[name] = {m: rrf_fuse([val_rank[best_fp][m], val_rank["B2"][m]], [1.0, w]) for m in truth.molecule_id}
        test_rank[name] = {m: rrf_fuse([test_rank[best_fp][m], test_rank["B2"][m]], [1.0, w]) for m in test_mols}
        tables[name] = score(val_rank[name])
        honest[name] = float(tables[name].loc["balanced", "MRR@25"])
    alts = ["B2", "fusion(B2 x2.0)"]
    gate_info = tune_gate(cp, truth, pool_keys, val_rank[best_fp], {n: val_rank[n] for n in alts}, val_b2_score)
    val_rank["gated B2"] = gate(val_rank[best_fp], val_rank[gate_info["alt"]], val_b2_score, gate_info["tau"])
    test_rank["gated B2"] = gate(test_rank[best_fp], test_rank[gate_info["alt"]], test_b2_score, gate_info["tau"])
    tables["gated B2"] = score(val_rank["gated B2"])
    honest["gated B2"] = gate_info["cross_fitted"]  # tau was tuned on validation: judge it out of sample
    gate_info["share_gated_test"] = round(float(np.mean([test_b2_score[m] >= gate_info["tau"] for m in test_mols])), 4)
    print("gate:", json.dumps(gate_info))
    before = cp.rank_table(val_rank[best_fp], truth, pool_keys).set_index("molecule_id")
    after = cp.rank_table(val_rank["gated B2"], truth, pool_keys).set_index("molecule_id")
    switched = [m for m in truth.molecule_id if val_b2_score[m] >= gate_info["tau"]]
    gate_log = pd.DataFrame({"molecule_id": switched, "cls": before.loc[switched, "cls"].to_numpy(),
                             "b2_score": [val_b2_score[m] for m in switched],
                             "rank_model": before.loc[switched, "rank"].to_numpy(),
                             "rank_gated": after.loc[switched, "rank"].to_numpy()})
    gate_log["effect"] = np.select([gate_log.rank_gated == gate_log.rank_model,
                                    (gate_log.rank_gated > 0) & ((gate_log.rank_model == 0)
                                                                 | (gate_log.rank_gated < gate_log.rank_model))],
                                   ["same", "helped"], "hurt")
    gate_log.to_csv(f"{OUT}/m7_results/val_gate_log.csv", index=False)
    print("gate switched", len(gate_log), "validation molecules:", gate_log.effect.value_counts().to_dict())
stamp("evaluation done")
```
